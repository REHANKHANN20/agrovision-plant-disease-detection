```python
# ================================================================
# POTATO CNN — REAL IMAGE TESTING
# ================================================================

import os
import numpy as np
import tensorflow as tf
from collections import Counter

from google.colab import drive
drive.mount('/content/drive')

PROJECT_DIR = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "6th Trained_Model",
    "potato_cnn_combined_best.keras"
)

# External real-image folder
TEST_DIR = os.path.join(
    PROJECT_DIR,
    "7th Test_Images",
    "all potato image",
    "extarnal potato dara from google and other sides"
)

class_names = ["Healthy", "Early_Blight", "Late_Blight"]

print("=" * 65)
print("POTATO CNN — REAL IMAGE TESTING")
print("=" * 65)

# Load model
model = tf.keras.models.load_model(MODEL_PATH)
print("✅ Model loaded")

# Find images recursively
image_files = []

for root, dirs, files in os.walk(TEST_DIR):
    for file in files:
        if file.lower().endswith((".jpg", ".jpeg", ".png", ".webp")):
            image_files.append(os.path.join(root, file))

print("Images found:", len(image_files))

if len(image_files) == 0:
    raise ValueError("❌ No images found. Check TEST_DIR path.")

# Predict
results = []

for path in image_files:

    img = tf.keras.utils.load_img(
        path,
        target_size=(224, 224),
        color_mode="rgb"
    )

    x = tf.keras.utils.img_to_array(img) / 255.0
    x = np.expand_dims(x, axis=0)

    prediction = model.predict(x, verbose=0)[0]

    pred_class = class_names[np.argmax(prediction)]
    confidence = float(np.max(prediction)) * 100

    results.append((path, pred_class, confidence))

# Display results
print("\n" + "=" * 65)
print("PREDICTIONS")
print("=" * 65)

for path, pred, conf in results:
    print(f"{os.path.basename(path):45s} → {pred:15s} {conf:.2f}%")

print("\n" + "=" * 65)
print("PREDICTION SUMMARY")
print("=" * 65)

summary = Counter(pred for _, pred, _ in results)

for cls in class_names:
    print(f"{cls:15s}: {summary[cls]}")

print("\n" + "=" * 65)
print("✅ POTATO REAL-IMAGE TESTING COMPLETED")
print("=" * 65)
```
