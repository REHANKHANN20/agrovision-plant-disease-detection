```python
# ==============================================================================
# STEP 1 — FINAL MULTI-PLANT DATASET STRUCTURE
# ==============================================================================

import os
from google.colab import drive

drive.mount('/content/drive')

PROJECT_DIR = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"

RAW_DIR = os.path.join(PROJECT_DIR, "1st Raw_Data")

plants = {
    "Potato": ["Internal_PlantVillage", "External_Natural"],
    "Tomato": ["Internal_PlantVillage", "External_Natural"],
    "Apple":  ["Internal_PlantVillage", "External_Natural"]
}

for plant, sources in plants.items():
    for source in sources:
        path = os.path.join(RAW_DIR, plant, source)
        os.makedirs(path, exist_ok=True)

print("=" * 70)
print("FINAL RAW DATASET STRUCTURE")
print("=" * 70)

for plant, sources in plants.items():
    print(f"\n{plant}")
    for source in sources:
        path = os.path.join(RAW_DIR, plant, source)
        print(f"  └── {source}")
        print(f"      {path}")

print("\n✅ Dataset structure ready.")

# ==============================================================================


# ==============================================================================
# STEP 2 — POTATO EXTERNAL NATURAL DATASET
# ==============================================================================

import os
import zipfile
import shutil
import requests
from tqdm.auto import tqdm

POTATO_EXTERNAL = os.path.join(
    RAW_DIR,
    "Potato",
    "External_Natural"
)

TEMP_DIR = "/content/potato_external_temp"
os.makedirs(TEMP_DIR, exist_ok=True)

os.makedirs(os.path.join(POTATO_EXTERNAL, "Healthy"), exist_ok=True)
os.makedirs(os.path.join(POTATO_EXTERNAL, "Early_Blight"), exist_ok=True)
os.makedirs(os.path.join(POTATO_EXTERNAL, "Late_Blight"), exist_ok=True)

print("=" * 70)
print("POTATO EXTERNAL DATASET DOWNLOAD")
print("=" * 70)

print("\nTarget:")
print("Healthy       : 1000 images")
print("Early_Blight  : 1000 images")
print("Late_Blight   : 1000 images")

print("\nSource:")
print("Zenodo — Irish Potato Imagery Dataset")

print("\n⚠️ Only a balanced subset will be retained.")

# ==============================================================================


# ==============================================================================
# STEP 2A — DOWNLOAD POTATO EXTERNAL ZIP FILES
# ==============================================================================

import requests
import os

zenodo_files = {
    "Healthy": "https://zenodo.org/records/8286529/files/healthy.zip?download=1",
    "Early_Blight": "https://zenodo.org/records/8286529/files/earlyblt.zip?download=1",
    "Late_Blight": "https://zenodo.org/records/8286529/files/lateblt.zip?download=1"
}

downloaded_files = {}

for class_name, url in zenodo_files.items():

    zip_path = os.path.join(
        TEMP_DIR,
        class_name + ".zip"
    )

    print(f"\nDownloading {class_name}...")

    response = requests.get(
        url,
        stream=True
    )

    response.raise_for_status()

    total_size = int(
        response.headers.get("content-length", 0)
    )

    with open(zip_path, "wb") as f:

        with tqdm(
            total=total_size,
            unit="B",
            unit_scale=True,
            desc=class_name
        ) as pbar:

            for chunk in response.iter_content(
                chunk_size=1024 * 1024
            ):

                if chunk:
                    f.write(chunk)
                    pbar.update(len(chunk))

    downloaded_files[class_name] = zip_path

    print(f"✅ {class_name} downloaded")

print("\n✅ All three Potato external ZIP files downloaded.")

# ==============================================================================



# ==============================================================================
# STEP 3 — EXTRACT BALANCED POTATO SUBSET
# ==============================================================================

import os
import zipfile
import shutil

TARGET_COUNT = 1000

class_mapping = {
    "Healthy": "Healthy",
    "Early_Blight": "Early_Blight",
    "Late_Blight": "Late_Blight"
}

for class_name, zip_path in downloaded_files.items():

    extract_dir = os.path.join(
        TEMP_DIR,
        "extracted_" + class_name
    )

    os.makedirs(extract_dir, exist_ok=True)

    print(f"\nExtracting {class_name}...")

    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall(extract_dir)

    image_files = []

    for root, dirs, files in os.walk(extract_dir):

        for file in files:

            if file.lower().endswith(
                (".jpg", ".jpeg", ".png", ".bmp", ".webp")
            ):
                image_files.append(
                    os.path.join(root, file)
                )

    print(
        f"Found {len(image_files):,} images"
    )

    selected = image_files[:TARGET_COUNT]

    destination = os.path.join(
        POTATO_EXTERNAL,
        class_name
    )

    for index, source_file in enumerate(selected):

        extension = os.path.splitext(source_file)[1]

        destination_file = os.path.join(
            destination,
            f"{class_name}_{index+1:04d}{extension}"
        )

        shutil.copy2(
            source_file,
            destination_file
        )

    print(
        f"✅ Copied {len(selected)} images → {destination}"
    )

print("\n" + "=" * 70)
print("POTATO EXTERNAL SUBSET COMPLETE")
print("=" * 70)


# ==============================================================================
# STEP 4 — POTATO EXTERNAL DATA VERIFICATION
# ==============================================================================

print("=" * 70)
print("POTATO EXTERNAL DATASET VERIFICATION")
print("=" * 70)

for class_name in [
    "Healthy",
    "Early_Blight",
    "Late_Blight"
]:

    folder = os.path.join(
        POTATO_EXTERNAL,
        class_name
    )

    count = len([
        f for f in os.listdir(folder)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp", ".webp")
        )
    ])

    print(
        f"{class_name:15s} : {count:4d} images"
    )

print("\nExpected:")
print("Healthy       : 1000")
print("Early_Blight  : 1000")
print("Late_Blight   : 1000")
```

```python
# ==============================================================================
# POTATO — NEW COMBINED RAW DATA VERIFICATION
# ==============================================================================

import os
from PIL import Image
from collections import Counter

print("=" * 75)
print("POTATO DATA VERIFICATION — INTERNAL + EXTERNAL")
print("=" * 75)

POTATO_RAW = os.path.join(
    PROJECT_DIR,
    "1st Raw_Data",
    "Potato"
)

sources = {
    "Internal_PlantVillage": os.path.join(
        POTATO_RAW, "Internal_PlantVillage"
    ),
    "External_Natural": os.path.join(
        POTATO_RAW, "External_Natural"
    )
}

classes = [
    "Healthy",
    "Early_Blight",
    "Late_Blight"
]

valid_extensions = (
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
)

verification = {}

for source_name, source_path in sources.items():

    print("\n" + "=" * 75)
    print(source_name)
    print("=" * 75)

    verification[source_name] = {}

    for class_name in classes:

        class_path = os.path.join(
            source_path,
            class_name
        )

        if not os.path.exists(class_path):
            print(f"❌ Missing: {class_name}")
            continue

        files = [
            f for f in os.listdir(class_path)
            if f.lower().endswith(valid_extensions)
        ]

        valid_images = 0
        corrupted_images = 0
        sizes = []

        for filename in files:

            filepath = os.path.join(
                class_path,
                filename
            )

            try:
                with Image.open(filepath) as img:
                    img.verify()

                with Image.open(filepath) as img:
                    sizes.append(img.size)

                valid_images += 1

            except Exception:
                corrupted_images += 1

        verification[source_name][class_name] = {
            "total": len(files),
            "valid": valid_images,
            "corrupted": corrupted_images,
            "sizes": sizes
        }

        print(
            f"{class_name:15s} | "
            f"Total: {len(files):4d} | "
            f"Valid: {valid_images:4d} | "
            f"Corrupt: {corrupted_images:3d}"
        )

print("\n" + "=" * 75)
print("POTATO VERIFICATION SUMMARY")
print("=" * 75)

for source_name in sources:

    print(f"\n{source_name}")

    total_valid = 0

    for class_name in classes:

        result = verification[source_name].get(
            class_name
        )

        if result:
            total_valid += result["valid"]

            print(
                f"  {class_name:15s}: "
                f"{result['valid']} valid"
            )

    print(f"  TOTAL VALID     : {total_valid}")

print("\n" + "=" * 75)
print("DATA VERIFICATION COMPLETED")
print("=" * 75)
```

```python
# ==============================================================================
# POTATO EXTERNAL EARLY_BLIGHT — FRESH DOWNLOAD + VERIFY + 300 IMAGE RECOVERY
# ==============================================================================

import os
import shutil
import zipfile
import requests
from io import BytesIO
from PIL import Image

print("=" * 75)
print("POTATO EXTERNAL EARLY_BLIGHT — FRESH RECOVERY")
print("=" * 75)

# ------------------------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------------------------

PROJECT_DIR = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"

EARLY_DIR = os.path.join(
    PROJECT_DIR,
    "1st Raw_Data",
    "Potato",
    "External_Natural",
    "Early_Blight"
)

TEMP_DIR = "/content/potato_early_blight_recovery"

os.makedirs(TEMP_DIR, exist_ok=True)

print("\nDestination:")
print(EARLY_DIR)

# ------------------------------------------------------------------------------
# 2. Remove ONLY corrupted Early_Blight folder
# ------------------------------------------------------------------------------

if os.path.exists(EARLY_DIR):
    print("\nRemoving old corrupted Early_Blight folder...")
    shutil.rmtree(EARLY_DIR)

os.makedirs(EARLY_DIR, exist_ok=True)

print("✅ Old Early_Blight folder removed.")
print("✅ New Early_Blight folder created.")

# ------------------------------------------------------------------------------
# 3. Download ONE official Zenodo Early_Blight ZIP
# ------------------------------------------------------------------------------

# Official Zenodo EARLYBLT_8.zip
ZIP_URL = "https://zenodo.org/records/17553016/files/EARLYBLT_8.zip"

ZIP_PATH = os.path.join(
    TEMP_DIR,
    "EARLYBLT_8.zip"
)

print("\n" + "=" * 75)
print("DOWNLOADING FRESH EARLY_BLIGHT DATA")
print("=" * 75)

print("\nSource:")
print(ZIP_URL)

print("\nDownloading EARLYBLT_8.zip...")

response = requests.get(
    ZIP_URL,
    stream=True,
    timeout=120
)

response.raise_for_status()

total_size = int(response.headers.get("content-length", 0))
downloaded = 0

with open(ZIP_PATH, "wb") as f:

    for chunk in response.iter_content(
        chunk_size=1024 * 1024
    ):

        if chunk:
            f.write(chunk)
            downloaded += len(chunk)

            if total_size:
                percent = downloaded * 100 / total_size

                print(
                    f"\rDownloaded: {percent:.1f}%",
                    end=""
                )

print("\n✅ ZIP download completed.")

# ------------------------------------------------------------------------------
# 4. Verify ZIP integrity
# ------------------------------------------------------------------------------

print("\nChecking ZIP integrity...")

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    bad_file = z.testzip()

    if bad_file is not None:
        raise RuntimeError(
            f"❌ ZIP integrity check failed: {bad_file}"
        )

print("✅ ZIP integrity verified.")

# ------------------------------------------------------------------------------
# 5. Find image files inside ZIP
# ------------------------------------------------------------------------------

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    image_files = [
        name
        for name in z.namelist()
        if name.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp", ".webp")
        )
    ]

    print("\nImages available in ZIP:")
    print(len(image_files))

    if len(image_files) < 300:
        raise ValueError(
            "❌ Fewer than 300 images available."
        )

    # --------------------------------------------------------------------------
    # 6. Verify images DIRECTLY from ZIP and copy first 300 valid images
    # --------------------------------------------------------------------------

    print("\n" + "=" * 75)
    print("VERIFYING IMAGES + SELECTING 300 VALID IMAGES")
    print("=" * 75)

    valid_count = 0
    checked_count = 0
    corrupt_count = 0

    for member in image_files:

        if valid_count >= 300:
            break

        checked_count += 1

        try:

            with z.open(member) as source:

                image_data = source.read()

            image = Image.open(
                BytesIO(image_data)
            )

            image.verify()

            # Re-open after verify
            image = Image.open(
                BytesIO(image_data)
            )

            image = image.convert("RGB")

            output_name = (
                f"Early_Blight_{valid_count + 1:04d}.jpg"
            )

            output_path = os.path.join(
                EARLY_DIR,
                output_name
            )

            image.save(
                output_path,
                format="JPEG",
                quality=95
            )

            valid_count += 1

        except Exception:
            corrupt_count += 1

    print("\nImages checked :", checked_count)
    print("Valid selected:", valid_count)
    print("Invalid found :", corrupt_count)

# ------------------------------------------------------------------------------
# 7. Final verification of saved 300 images
# ------------------------------------------------------------------------------

print("\n" + "=" * 75)
print("FINAL EARLY_BLIGHT VERIFICATION")
print("=" * 75)

final_files = [
    f for f in os.listdir(EARLY_DIR)
    if f.lower().endswith(
        (".jpg", ".jpeg", ".png", ".bmp", ".webp")
    )
]

final_valid = 0
final_corrupt = 0

for filename in final_files:

    filepath = os.path.join(
        EARLY_DIR,
        filename
    )

    try:

        with Image.open(filepath) as img:
            img.verify()

        final_valid += 1

    except Exception:
        final_corrupt += 1

print("Final files     :", len(final_files))
print("Final valid     :", final_valid)
print("Final corrupt   :", final_corrupt)

# ------------------------------------------------------------------------------
# 8. Final result
# ------------------------------------------------------------------------------

print("\n" + "=" * 75)
print("EARLY_BLIGHT RECOVERY RESULT")
print("=" * 75)

if final_valid == 300 and final_corrupt == 0:

    print("✅ SUCCESS")
    print("✅ 300 valid Early_Blight images recovered.")
    print("✅ Images verified after saving.")
    print("✅ Potato Early_Blight external class is READY.")

else:

    print("❌ RECOVERY NOT COMPLETE")
    print("Please inspect the output before continuing.")

print("\nPath:")
print(EARLY_DIR)

print("\n" + "=" * 75)
print("POTATO EARLY_BLIGHT RECOVERY COMPLETED")
print("=" * 75)
```

```python
# ==============================================================================
# POTATO — FINAL RAW DATASET VERIFICATION & LOCK
# ==============================================================================

import os
from PIL import Image

print("=" * 75)
print("POTATO — FINAL RAW DATASET VERIFICATION")
print("=" * 75)

POTATO_RAW = os.path.join(
    PROJECT_DIR,
    "1st Raw_Data",
    "Potato"
)

sources = {
    "Internal_PlantVillage": os.path.join(
        POTATO_RAW, "Internal_PlantVillage"
    ),
    "External_Natural": os.path.join(
        POTATO_RAW, "External_Natural"
    )
}

classes = [
    "Healthy",
    "Early_Blight",
    "Late_Blight"
]

valid_extensions = (
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
)

all_valid = True
total_valid = 0

print("\nChecking all 6 source/class groups...\n")

for source_name, source_path in sources.items():

    print("-" * 75)
    print(source_name)
    print("-" * 75)

    for class_name in classes:

        class_path = os.path.join(
            source_path,
            class_name
        )

        if not os.path.exists(class_path):
            print(f"❌ {class_name:15s} | FOLDER MISSING")
            all_valid = False
            continue

        files = [
            f for f in os.listdir(class_path)
            if f.lower().endswith(valid_extensions)
        ]

        valid = 0
        corrupt = 0

        for filename in files:

            filepath = os.path.join(
                class_path,
                filename
            )

            try:
                with Image.open(filepath) as img:
                    img.verify()

                valid += 1

            except Exception:
                corrupt += 1

        total_valid += valid

        status = "✅ VALID" if corrupt == 0 and valid > 0 else "❌ PROBLEM"

        print(
            f"{class_name:15s} | "
            f"Images: {valid:4d} | "
            f"Corrupt: {corrupt:3d} | "
            f"{status}"
        )

        if corrupt > 0 or valid == 0:
            all_valid = False

# ------------------------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------------------------

print("\n" + "=" * 75)
print("FINAL POTATO DATASET SUMMARY")
print("=" * 75)

print("\nInternal / PlantVillage")
print("Healthy       : 152")
print("Early_Blight  : 1000")
print("Late_Blight   : 1000")
print("Total         : 2152")

print("\nExternal / Natural")
print("Healthy       : 1000")
print("Early_Blight  : 300")
print("Late_Blight   : 1000")
print("Total         : 2300")

print("\nCombined valid images :", total_valid)

# ------------------------------------------------------------------------------
# LOCK DECISION
# ------------------------------------------------------------------------------

print("\n" + "=" * 75)
print("POTATO RAW DATASET LOCK STATUS")
print("=" * 75)

if all_valid and total_valid == 4452:

    print("✅ ALL 6 DATA GROUPS VERIFIED")
    print("✅ NO CORRUPTED IMAGES FOUND")
    print("✅ INTERNAL + EXTERNAL DATASET READY")
    print("🔒 POTATO RAW DATASET LOCKED")
    print("\n➡️ NEXT: POTATO DATA PREPROCESSING")

else:

    print("❌ DATASET NOT LOCKED")
    print("⚠️ Fix the reported problem before preprocessing.")

print("=" * 75)
```
