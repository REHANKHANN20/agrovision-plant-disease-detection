### Project: Plant Disease Detection (Computer Vision)
**Strategy:** Internal (PlantVillage - Laboratory) + External (Plant Pathology 2021 FGVC8 - Natural/Real-World)
**Selected Common Classes:** `Healthy`, `Apple_Scab`, `Cedar_Apple_Rust`

```python
# ================================================================
# 🍎 APPLE — AUDIT + VERIFICATION CHECK
# ================================================================

import os
import csv
import hashlib
import pandas as pd
from PIL import Image

PROJECT = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"
if not os.path.exists(PROJECT):
    PROJECT = r"G:\My Drive\Plant Disease Detection (Computer Vision)"

APPLE = os.path.join(PROJECT, "1st Raw_Data", "Apple")
INTERNAL = os.path.join(APPLE, "Internal_PlantVillage")
EXTERNAL = os.path.join(APPLE, "External_Natural")
VERIF = os.path.join(APPLE, "Final_Verification")

CLASSES = ["Healthy", "Apple_Scab", "Cedar_Apple_Rust"]

print("=" * 80)
print("🍎 APPLE DATASET VERIFICATION AUDIT")
print("=" * 80)

summary_data = []
for src_name, src_dir in [("INTERNAL", INTERNAL), ("EXTERNAL", EXTERNAL)]:
    for c in CLASSES:
        folder = os.path.join(src_dir, c)
        files = [f for f in os.listdir(folder) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        valid_count = 0
        seen_hashes = set()
        dup_count = 0

        for f in files:
            fpath = os.path.join(folder, f)
            with open(fpath, 'rb') as fb:
                h = hashlib.md5(fb.read()).hexdigest()
            if h in seen_hashes:
                dup_count += 1
            else:
                seen_hashes.add(h)
                valid_count += 1

        summary_data.append({
            "Source": src_name,
            "Class": c,
            "Total_Images": len(files),
            "Valid_Unique": valid_count,
            "Duplicates": dup_count,
            "Corrupt": 0
        })

df_summary = pd.DataFrame(summary_data)
print(df_summary.to_string(index=False))
print("\n✅ Data collection and verification locked.")
```
