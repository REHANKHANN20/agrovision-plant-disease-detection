```python
# ================================================================
# TOMATO — AUDIT + CLEAN + FRESH RAW DATA SETUP
# ================================================================

import os
import shutil
import hashlib
from PIL import Image
from google.colab import drive

drive.mount("/content/drive")

# ------------------------------------------------
# PATHS
# ------------------------------------------------
PROJECT = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"
TOMATO = os.path.join(PROJECT, "1st Raw_Data", "Tomato")

INTERNAL = os.path.join(TOMATO, "Internal_PlantVillage")
EXTERNAL = os.path.join(TOMATO, "External_Natural")

CLASSES = ["Healthy", "Early_Blight", "Late_Blight"]

TARGET = 300

# ------------------------------------------------
# AUDIT FUNCTION
# ------------------------------------------------
def audit_dataset(base_path):

    report = {}

    for cls in CLASSES:

        folder = os.path.join(base_path, cls)

        valid = 0
        corrupt = 0
        duplicates = 0
        hashes = set()

        if os.path.exists(folder):

            for root, _, files in os.walk(folder):

                for file in files:

                    if not file.lower().endswith(
                        (".jpg", ".jpeg", ".png", ".webp")
                    ):
                        continue

                    path = os.path.join(root, file)

                    try:
                        with Image.open(path) as img:
                            img.verify()

                        with open(path, "rb") as f:
                            h = hashlib.md5(f.read()).hexdigest()

                        if h in hashes:
                            duplicates += 1
                        else:
                            hashes.add(h)

                        valid += 1

                    except:
                        corrupt += 1

        report[cls] = (valid, corrupt, duplicates)

    return report


# ================================================================
# 1. CURRENT INTERNAL AUDIT
# ================================================================

print("=" * 70)
print("TOMATO — CURRENT INTERNAL AUDIT")
print("=" * 70)

internal = audit_dataset(INTERNAL)

for cls, (valid, corrupt, dup) in internal.items():
    print(
        f"{cls:15s} | Valid: {valid:4d} | "
        f"Corrupt: {corrupt:3d} | Duplicate: {dup:3d}"
    )


# ================================================================
# 2. CURRENT EXTERNAL AUDIT
# ================================================================

print("\n" + "=" * 70)
print("TOMATO — CURRENT EXTERNAL AUDIT")
print("=" * 70)

external = audit_dataset(EXTERNAL)

for cls, (valid, corrupt, dup) in external.items():
    print(
        f"{cls:15s} | Valid: {valid:4d} | "
        f"Corrupt: {corrupt:3d} | Duplicate: {dup:3d}"
    )


# ================================================================
# 3. REMOVE OLD TOMATO DATA ONLY
# ================================================================

print("\n" + "=" * 70)
print("CLEANING TOMATO RAW DATA")
print("=" * 70)

if os.path.exists(TOMATO):
    shutil.rmtree(TOMATO)

os.makedirs(INTERNAL, exist_ok=True)
os.makedirs(EXTERNAL, exist_ok=True)

for cls in CLASSES:
    os.makedirs(os.path.join(INTERNAL, cls), exist_ok=True)
    os.makedirs(os.path.join(EXTERNAL, cls), exist_ok=True)

print("✅ Old Tomato data removed")
print("✅ Fresh Tomato folders created")
print("🔒 Potato untouched")


# ================================================================
# 4. FINAL DOWNLOAD PLAN
# ================================================================

print("\n" + "=" * 70)
print("TOMATO RAW DATA DOWNLOAD PLAN")
print("=" * 70)

print("""
INTERNAL — PlantVillage
    Healthy       → 300
    Early_Blight  → 300
    Late_Blight   → 300

EXTERNAL — Natural/Field
    Healthy       → 300
    Early_Blight  → 300
    Late_Blight   → 300

TOTAL TARGET = 1,800 images

Rule:
✅ Only required classes
✅ Only 300 images per class
✅ No augmented dataset
✅ No unnecessary large download
✅ Every image verified after saving
""")

print("=" * 70)
print("TOMATO AUDIT + CLEANUP COMPLETED")
print("=" * 70)

print("➡️ NEXT:")
print("Download/select 300 valid images per class")
print("→ Verify")
print("→ Final Tomato 6-group verification")



# ================================================================
# 🍅 TOMATO — RAW DATA FINAL AUDIT + 300/CLASS SELECTION
# ================================================================

import os
import shutil
import hashlib
from PIL import Image
from collections import Counter

from google.colab import drive
drive.mount("/content/drive")

# ------------------------------------------------
# 1. PATHS
# ------------------------------------------------

PROJECT = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"

TOMATO = os.path.join(PROJECT, "1st Raw_Data", "Tomato")

INTERNAL = os.path.join(TOMATO, "Internal_PlantVillage")
EXTERNAL = os.path.join(TOMATO, "External_Natural")

CLASSES = ["Healthy", "Early_Blight", "Late_Blight"]

TARGET = 300

# ------------------------------------------------
# 2. AUDIT FUNCTION
# ------------------------------------------------

def audit_dataset(base_path):

    report = {}

    for cls in CLASSES:

        folder = os.path.join(base_path, cls)

        valid = []
        corrupt = 0
        hashes = set()
        duplicates = 0

        if os.path.exists(folder):

            for root, _, files in os.walk(folder):

                for file in files:

                    if not file.lower().endswith(
                        (".jpg", ".jpeg", ".png", ".webp")
                    ):
                        continue

                    path = os.path.join(root, file)

                    try:

                        with Image.open(path) as img:
                            img.verify()

                        with open(path, "rb") as f:
                            file_hash = hashlib.md5(
                                f.read()
                            ).hexdigest()

                        if file_hash in hashes:
                            duplicates += 1
                            continue

                        hashes.add(file_hash)
                        valid.append(path)

                    except:
                        corrupt += 1

        report[cls] = {
            "valid": valid,
            "corrupt": corrupt,
            "duplicates": duplicates
        }

    return report


# ------------------------------------------------
# 3. AUDIT INTERNAL
# ------------------------------------------------

print("=" * 70)
print("🍅 TOMATO — INTERNAL AUDIT")
print("=" * 70)

internal_report = audit_dataset(INTERNAL)

for cls in CLASSES:

    r = internal_report[cls]

    print(
        f"{cls:15s} | "
        f"Valid: {len(r['valid']):4d} | "
        f"Corrupt: {r['corrupt']:3d} | "
        f"Duplicate: {r['duplicates']:3d}"
    )


# ------------------------------------------------
# 4. AUDIT EXTERNAL
# ------------------------------------------------

print("\n" + "=" * 70)
print("🍅 TOMATO — EXTERNAL AUDIT")
print("=" * 70)

external_report = audit_dataset(EXTERNAL)

for cls in CLASSES:

    r = external_report[cls]

    print(
        f"{cls:15s} | "
        f"Valid: {len(r['valid']):4d} | "
        f"Corrupt: {r['corrupt']:3d} | "
        f"Duplicate: {r['duplicates']:3d}"
    )


# ------------------------------------------------
# 5. CHECK AVAILABLE DATA
# ------------------------------------------------

print("\n" + "=" * 70)
print("AVAILABLE VALID DATA")
print("=" * 70)

for cls in CLASSES:

    internal_count = len(internal_report[cls]["valid"])
    external_count = len(external_report[cls]["valid"])

    print(
        f"{cls:15s} | "
        f"Internal: {internal_count:4d} | "
        f"External: {external_count:4d}"
    )


# ------------------------------------------------
# 6. VERIFY 300/CLASS REQUIREMENT
# ------------------------------------------------

print("\n" + "=" * 70)
print("300 IMAGE REQUIREMENT CHECK")
print("=" * 70)

ready = True

for cls in CLASSES:

    i = len(internal_report[cls]["valid"])
    e = len(external_report[cls]["valid"])

    if i < TARGET:
        print(f"⚠️ Internal {cls}: only {i} available")

    else:
        print(f"✅ Internal {cls}: {i} available → 300 required")

    if e < TARGET:
        print(f"⚠️ External {cls}: only {e} available")

    else:
        print(f"✅ External {cls}: {e} available → 300 required")

    if i < TARGET or e < TARGET:
        ready = False


# ------------------------------------------------
# 7. STOP IF SOURCE DATA IS INSUFFICIENT
# ------------------------------------------------

if not ready:

    print("\n" + "=" * 70)
    print("⚠️ TOMATO DATA NOT READY")
    print("=" * 70)

    print("""
At least one class/source has fewer than 300 valid images.

Do NOT train the model yet.

Next action:
→ Download/collect additional RAW images
→ Re-run this audit
→ Continue only when every class has >=300 valid images
""")

else:

    # ------------------------------------------------
    # 8. CREATE FINAL 300-IMAGE DATASET
    # ------------------------------------------------

    print("\n" + "=" * 70)
    print("CREATING FINAL 300/CLASS TOMATO DATA")
    print("=" * 70)

    # Clean only selected final folders
    for base in [INTERNAL, EXTERNAL]:

        for cls in CLASSES:

            folder = os.path.join(base, cls)

            os.makedirs(folder, exist_ok=True)

            # Remove current contents
            for item in os.listdir(folder):

                item_path = os.path.join(folder, item)

                if os.path.isfile(item_path):
                    os.remove(item_path)

                elif os.path.isdir(item_path):
                    shutil.rmtree(item_path)

    # ------------------------------------------------
    # 9. COPY EXACTLY 300 IMAGES
    # ------------------------------------------------

    def select_images(report, base_path):

        for cls in CLASSES:

            source_images = report[cls]["valid"]

            selected = source_images[:TARGET]

            destination = os.path.join(base_path, cls)

            for index, source in enumerate(selected, start=1):

                extension = os.path.splitext(source)[1].lower()

                destination_file = os.path.join(
                    destination,
                    f"{cls}_{index:03d}{extension}"
                )

                shutil.copy2(source, destination_file)

            print(
                f"✅ {os.path.basename(base_path):20s} "
                f"{cls:15s} → {len(selected)} images"
            )


    select_images(internal_report, INTERNAL)
    select_images(external_report, EXTERNAL)


    # ------------------------------------------------
    # 10. FINAL VERIFICATION
    # ------------------------------------------------

    print("\n" + "=" * 70)
    print("FINAL TOMATO DATA VERIFICATION")
    print("=" * 70)

    final_ok = True

    for source_name, base in [
        ("Internal", INTERNAL),
        ("External", EXTERNAL)
    ]:

        for cls in CLASSES:

            folder = os.path.join(base, cls)

            count = len([
                f for f in os.listdir(folder)
                if f.lower().endswith(
                    (".jpg", ".jpeg", ".png", ".webp")
                )
            ])

            print(
                f"{source_name:10s} | "
                f"{cls:15s} | "
                f"{count} images"
            )

            if count != TARGET:
                final_ok = False


    # ------------------------------------------------
    # 11. FINAL STATUS
    # ------------------------------------------------

    print("\n" + "=" * 70)

    if final_ok:

        print("✅ TOMATO RAW DATA READY")
        print("=" * 70)

        print("""
Internal:
    Healthy       = 300
    Early_Blight  = 300
    Late_Blight   = 300

External:
    Healthy       = 300
    Early_Blight  = 300
    Late_Blight   = 300

TOTAL = 1,800 images

🔒 Internal + External kept separate
🔒 300 images/class/source
🔒 No augmented dataset
🔒 Corrupt images excluded
🔒 Exact duplicates excluded

NEXT → TOMATO FINAL 6-GROUP VERIFICATION
""")

    else:

        print("❌ FINAL VERIFICATION FAILED")
        print("Check the folder counts before continuing.")
```

```python
# ================================================================
# 🍅 TOMATO — DOWNLOAD 300 IMAGES / CLASS
# INTERNAL + EXTERNAL
# ================================================================

import os
import shutil
import zipfile
import requests
import random
from PIL import Image
from google.colab import drive

drive.mount("/content/drive")

# ================================================================
# PATHS
# ================================================================

PROJECT = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"
TOMATO = os.path.join(PROJECT, "1st Raw_Data", "Tomato")

INTERNAL = os.path.join(TOMATO, "Internal_PlantVillage")
EXTERNAL = os.path.join(TOMATO, "External_Natural")

CLASSES = ["Healthy", "Early_Blight", "Late_Blight"]
TARGET = 300

TMP = "/content/tomato_download"

os.makedirs(TMP, exist_ok=True)

# ================================================================
# HELPER — SAVE VALID IMAGES
# ================================================================

def save_valid_images(source_folder, destination, target):

    candidates = []

    for root, _, files in os.walk(source_folder):
        for f in files:

            if f.lower().endswith((".jpg", ".jpeg", ".png", ".webp")):
                candidates.append(os.path.join(root, f))

    random.shuffle(candidates)

    saved = 0

    for src in candidates:

        if saved >= target:
            break

        try:
            with Image.open(src) as img:
                img.verify()

            ext = os.path.splitext(src)[1].lower()

            dst = os.path.join(
                destination,
                f"Tomato_{os.path.basename(destination)}_{saved+1:03d}{ext}"
            )

            shutil.copy2(src, dst)
            saved += 1

        except:
            continue

    return saved


# ================================================================
# 1. DOWNLOAD PLANTVILLAGE
# ================================================================

print("=" * 70)
print("🍅 TOMATO INTERNAL — PLANTVILLAGE")
print("=" * 70)

KAGGLE_DATASET = "emmarex/plantdisease"

# Install Kaggle
os.system("pip -q install kaggle")

# Check Kaggle authentication
if not os.path.exists("/root/.kaggle/kaggle.json"):
    print("""
⚠️ Kaggle API key is required.

In Colab:
1. Open Kaggle → Account
2. Create New API Token
3. Upload kaggle.json when prompted
""")

    from google.colab import files
    uploaded = files.upload()

    os.makedirs("/root/.kaggle", exist_ok=True)

    shutil.copy(
        list(uploaded.keys())[0],
        "/root/.kaggle/kaggle.json"
    )

    os.chmod("/root/.kaggle/kaggle.json", 0o600)

# Download PlantVillage
PV_DIR = os.path.join(TMP, "plantvillage")

if not os.path.exists(PV_DIR):

    os.makedirs(PV_DIR, exist_ok=True)

    print("Downloading PlantVillage...")
    os.system(
        f'kaggle datasets download -d {KAGGLE_DATASET} '
        f'-p "{PV_DIR}" --unzip'
    )

print("✅ PlantVillage available")


# ================================================================
# 2. FIND TOMATO CLASSES
# ================================================================

print("\nSearching Tomato classes...")

def find_class_folder(base, keywords):

    for root, dirs, files in os.walk(base):

        name = os.path.basename(root).lower()

        if all(k.lower() in name for k in keywords):

            image_count = sum(
                1 for f in files
                if f.lower().endswith(
                    (".jpg", ".jpeg", ".png", ".webp")
                )
            )

            if image_count > 0:
                return root

    return None


tomato_internal_sources = {

    "Healthy": find_class_folder(
        PV_DIR, ["tomato", "healthy"]
    ),

    "Early_Blight": find_class_folder(
        PV_DIR, ["tomato", "early", "blight"]
    ),

    "Late_Blight": find_class_folder(
        PV_DIR, ["tomato", "late", "blight"]
    )
}


# ================================================================
# 3. SAVE 300 INTERNAL IMAGES
# ================================================================

print("\nSelecting 300 images/class...")

for cls in CLASSES:

    source = tomato_internal_sources[cls]
    destination = os.path.join(INTERNAL, cls)

    os.makedirs(destination, exist_ok=True)

    if source is None:
        print(f"❌ Internal {cls} source not found")
        continue

    saved = save_valid_images(
        source,
        destination,
        TARGET
    )

    print(
        f"Internal {cls:15s} → {saved} images"
    )


# ================================================================
# 4. DOWNLOAD MENDELEY ORIGINAL DATASET
# ================================================================

print("\n" + "=" * 70)
print("🍅 TOMATO EXTERNAL — MENDELEY ORIGINAL DATA")
print("=" * 70)

# Original Dataset.zip
MENDELEY_URL = (
    "https://data.mendeley.com/public-files/"
    "datasets/9jxvtgh325/1/files/"
)

print("""
Source:
Mendeley Data — Tomato Leaf and Fruit Disease Image Dataset

Using:
✅ Original Dataset
❌ NOT Augmented Dataset
""")

# Try API page extraction to locate downloadable original ZIP
page_url = "https://data.mendeley.com/datasets/9jxvtgh325/1"

html = requests.get(
    page_url,
    timeout=60,
    headers={"User-Agent": "Mozilla/5.0"}
).text

import re

zip_links = re.findall(
    r'https?[^"\']+Original[^"\']+\.zip[^"\']*',
    html,
    flags=re.I
)

if not zip_links:

    print("""
⚠️ Direct Mendeley ZIP URL was not exposed by the page.

The official dataset is reachable, but Mendeley's
download endpoint may require a browser/session.

Open:
https://data.mendeley.com/datasets/9jxvtgh325/1

Download ONLY:
Original Dataset.zip

NOT:
Augmented Dataset.zip
""")

else:

    original_url = zip_links[0].replace("\\u0026", "&")

    ZIP_PATH = os.path.join(
        TMP,
        "tomato_original.zip"
    )

    print("Downloading Original Dataset...")

    response = requests.get(
        original_url,
        stream=True,
        timeout=120,
        headers={"User-Agent": "Mozilla/5.0"}
    )

    response.raise_for_status()

    with open(ZIP_PATH, "wb") as f:

        for chunk in response.iter_content(
            chunk_size=1024 * 1024
        ):

            if chunk:
                f.write(chunk)

    print("✅ Original Dataset downloaded")

    # Extract
    EXT_DIR = os.path.join(
        TMP,
        "mendeley_original"
    )

    os.makedirs(EXT_DIR, exist_ok=True)

    print("Extracting...")

    with zipfile.ZipFile(
        ZIP_PATH,
        "r"
    ) as z:

        z.extractall(EXT_DIR)

    print("✅ Extraction completed")


# ================================================================
# 5. FIND MENDELEY CLASSES
# ================================================================

EXT_SOURCE = os.path.join(
    TMP,
    "mendeley_original"
)

if os.path.exists(EXT_SOURCE):

    external_sources = {

        "Healthy": find_class_folder(
            EXT_SOURCE, ["healthy", "leaf"]
        ),

        "Early_Blight": find_class_folder(
            EXT_SOURCE, ["early", "blight", "leaf"]
        ),

        "Late_Blight": find_class_folder(
            EXT_SOURCE, ["late", "blight", "leaf"]
        )
    }

    # ============================================================
    # 6. SAVE 300 EXTERNAL IMAGES
    # ============================================================

    print("\nSelecting 300 natural images/class...")

    for cls in CLASSES:

        source = external_sources[cls]
        destination = os.path.join(EXTERNAL, cls)

        os.makedirs(destination, exist_ok=True)

        if source is None:
            print(
                f"❌ External {cls} source not found"
            )
            continue

        saved = save_valid_images(
            source,
            destination,
            TARGET
        )

        print(
            f"External {cls:15s} → {saved} images"
        )


# ================================================================
# 7. FINAL DOWNLOAD COUNT
# ================================================================

print("\n" + "=" * 70)
print("🍅 TOMATO RAW DATA DOWNLOAD SUMMARY")
print("=" * 70)

for source_name, base in [
    ("Internal", INTERNAL),
    ("External", EXTERNAL)
]:

    print(f"\n{source_name}")

    for cls in CLASSES:

        folder = os.path.join(base, cls)

        count = 0

        if os.path.exists(folder):

            count = len([
                f for f in os.listdir(folder)
                if f.lower().endswith(
                    (".jpg", ".jpeg", ".png", ".webp")
                )
            ])

        print(
            f"{cls:15s}: {count}"
        )

print("""
================================================================
TARGET
================================================================

Internal:
Healthy       = 300
Early_Blight  = 300
Late_Blight   = 300

External:
Healthy       = 300
Early_Blight  = 300
Late_Blight  = 300

TOTAL TARGET = 1,800 images

NEXT → FINAL TOMATO 6-GROUP VERIFICATION
================================================================
""")
```

```python
# ================================================================
# 🍅 TOMATO EXTERNAL — DRIVE → DRIVE VERIFICATION
# NO UPLOAD • NO ZIP • NO EXTRACTION
# ================================================================

import os
import csv
import hashlib
import shutil
from PIL import Image
from google.colab import drive

# ================================================================
# 1. MOUNT GOOGLE DRIVE
# ================================================================

drive.mount("/content/drive")

# ================================================================
# 2. PATHS
# ================================================================

PROJECT = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"

RAW = os.path.join(
    PROJECT,
    "1st Raw_Data",
    "Tomato",
    "External_Natural"
)

VERIFIED = os.path.join(
    RAW,
    "Verified_300"
)

REPORT = os.path.join(
    RAW,
    "Verification_Report"
)

# ================================================================
# 3. REQUIRED CLASSES
# ================================================================

CLASSES = {
    "Healthy": [
        "Healthy",
        "Healthy Leaf"
    ],

    "Early_Blight": [
        "Early_Blight",
        "Early Blight",
        "Early Blight Leaf"
    ],

    "Late_Blight": [
        "Late_Blight",
        "Late Blight",
        "Late Blight Leaf"
    ]
}

TARGET = 300

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp"
)

# ================================================================
# 4. CHECK RAW FOLDER
# ================================================================

print("=" * 75)
print("🍅 TOMATO EXTERNAL — DRIVE VERIFICATION")
print("=" * 75)

print("\nRAW DATA LOCATION:")
print(RAW)

if not os.path.exists(RAW):
    raise Exception(
        "\n❌ External_Natural folder not found.\n"
        "Check your Google Drive path."
    )

print("\n✅ External_Natural folder found.")

# ================================================================
# 5. SHOW ACTUAL FOLDERS
# ================================================================

print("\n" + "=" * 75)
print("📁 FOLDERS FOUND IN EXTERNAL_NATURAL")
print("=" * 75)

actual_folders = []

for item in os.listdir(RAW):

    full_path = os.path.join(RAW, item)

    if os.path.isdir(full_path):
        actual_folders.append(item)

        print("→", item)

# ================================================================
# 6. FIND REQUIRED FOLDERS
# ================================================================

def find_folder(possible_names):

    # First try exact match
    for name in possible_names:

        path = os.path.join(RAW, name)

        if os.path.isdir(path):
            return path

    # Then case-insensitive match
    for actual in actual_folders:

        for name in possible_names:

            if actual.lower() == name.lower():
                return os.path.join(RAW, actual)

    return None


SOURCE_FOLDERS = {}

for cls, names in CLASSES.items():

    folder = find_folder(names)

    if folder is None:

        print(
            f"❌ {cls}: folder NOT FOUND"
        )

    else:

        SOURCE_FOLDERS[cls] = folder

        print(
            f"✅ {cls:15s} → {os.path.basename(folder)}"
        )

# ================================================================
# 7. REQUIRED FOLDER CHECK
# ================================================================

missing = [
    cls for cls in CLASSES
    if cls not in SOURCE_FOLDERS
]

if missing:

    raise Exception(
        "\n❌ Required folder(s) missing: "
        + ", ".join(missing)
        + "\n\nDo not continue."
    )

# ================================================================
# 8. CREATE REPORT FOLDER
# ================================================================

os.makedirs(REPORT, exist_ok=True)

# ================================================================
# 9. SCAN + VERIFY
# ================================================================

print("\n" + "=" * 75)
print("🔍 SCANNING AND VERIFYING IMAGES")
print("=" * 75)

verification_results = {}

for cls in CLASSES:

    source = SOURCE_FOLDERS[cls]

    print("\n" + "-" * 75)
    print(f"CLASS: {cls}")
    print(f"SOURCE: {source}")
    print("-" * 75)

    image_files = []

    # ------------------------------------------------------------
    # FIND IMAGE FILES
    # ------------------------------------------------------------

    for root, _, files_list in os.walk(source):

        for filename in files_list:

            if filename.lower().endswith(IMAGE_EXTENSIONS):

                image_files.append(
                    os.path.join(root, filename)
                )

    print(
        f"Total image files found: {len(image_files)}"
    )

    # ------------------------------------------------------------
    # VERIFY EACH IMAGE
    # ------------------------------------------------------------

    valid_files = []
    corrupt_files = []

    format_count = {}
    dimension_count = {}

    hashes = {}
    duplicate_files = []

    csv_rows = []

    for index, path in enumerate(
        image_files,
        start=1
    ):

        filename = os.path.basename(path)

        extension = os.path.splitext(
            filename
        )[1].lower()

        try:

            # ----------------------------------------------------
            # IMAGE VALIDATION
            # ----------------------------------------------------

            with Image.open(path) as img:

                img.verify()

            # Reopen because verify() closes image state
            with Image.open(path) as img:

                image_format = img.format
                width, height = img.size
                mode = img.mode

            # ----------------------------------------------------
            # HASH
            # ----------------------------------------------------

            md5 = hashlib.md5()

            with open(path, "rb") as f:

                for chunk in iter(
                    lambda: f.read(1024 * 1024),
                    b""
                ):

                    md5.update(chunk)

            file_hash = md5.hexdigest()

            # ----------------------------------------------------
            # DUPLICATE CHECK
            # ----------------------------------------------------

            if file_hash in hashes:

                duplicate_files.append(
                    path
                )

                status = "DUPLICATE"

            else:

                hashes[file_hash] = path

                valid_files.append(path)

                status = "VALID"

            # ----------------------------------------------------
            # FORMAT COUNT
            # ----------------------------------------------------

            format_count[image_format] = (
                format_count.get(
                    image_format,
                    0
                ) + 1
            )

            # ----------------------------------------------------
            # DIMENSION COUNT
            # ----------------------------------------------------

            dimension = f"{width}x{height}"

            dimension_count[dimension] = (
                dimension_count.get(
                    dimension,
                    0
                ) + 1
            )

            csv_rows.append([
                cls,
                filename,
                extension,
                image_format,
                width,
                height,
                mode,
                file_hash,
                status,
                path
            ])

        except Exception as e:

            corrupt_files.append(path)

            csv_rows.append([
                cls,
                filename,
                extension,
                "UNKNOWN",
                "",
                "",
                "",
                "",
                "CORRUPT",
                str(e)
            ])

    # ------------------------------------------------------------
    # RESULTS
    # ------------------------------------------------------------

    verification_results[cls] = {
        "total": len(image_files),
        "valid": len(valid_files),
        "corrupt": len(corrupt_files),
        "duplicates": len(duplicate_files),
        "unique_valid": len(valid_files),
        "formats": format_count,
        "dimensions": dimension_count,
        "valid_files": valid_files,
        "corrupt_files": corrupt_files,
        "duplicate_files": duplicate_files,
        "csv_rows": csv_rows
    }

    print(
        f"Total       : {len(image_files)}"
    )

    print(
        f"Valid       : {len(valid_files)}"
    )

    print(
        f"Corrupt     : {len(corrupt_files)}"
    )

    print(
        f"Duplicates  : {len(duplicate_files)}"
    )

    print(
        f"Unique valid: {len(valid_files)}"
    )

    print("\nFormats:")

    for fmt, count in format_count.items():

        print(
            f"  {fmt:10s} : {count}"
        )

    print("\nDimensions:")

    for dim, count in sorted(
        dimension_count.items(),
        key=lambda x: x[1],
        reverse=True
    )[:10]:

        print(
            f"  {dim:12s} : {count}"
        )

    # ============================================================
    # SAVE CSV REPORT
    # ============================================================

    report_file = os.path.join(
        REPORT,
        f"{cls}_verification.csv"
    )

    with open(
        report_file,
        "w",
        newline="",
        encoding="utf-8"
    ) as f:

        writer = csv.writer(f)

        writer.writerow([
            "Class",
            "Filename",
            "Extension",
            "Format",
            "Width",
            "Height",
            "Mode",
            "MD5",
            "Status",
            "Path"
        ])

        writer.writerows(csv_rows)

    print(
        f"\n📄 Report saved: {report_file}"
    )

# ================================================================
# 10. CREATE EXACT 300 VERIFIED DATASET
# ================================================================

print("\n" + "=" * 75)
print("📦 CREATING EXACT 300 VERIFIED IMAGES / CLASS")
print("=" * 75)

selection_ok = True

for cls in CLASSES:

    results = verification_results[cls]

    valid_files = results["valid_files"]

    # ------------------------------------------------------------
    # CHECK 300
    # ------------------------------------------------------------

    if len(valid_files) < TARGET:

        print(
            f"\n❌ {cls}: only "
            f"{len(valid_files)} unique valid images."
        )

        selection_ok = False

        continue

    # ------------------------------------------------------------
    # CREATE CLASS OUTPUT
    # ------------------------------------------------------------

    destination = os.path.join(
        VERIFIED,
        cls
    )

    os.makedirs(
        destination,
        exist_ok=True
    )

    # ------------------------------------------------------------
    # IMPORTANT:
    # Select first 300 valid unique images
    # Deterministic selection
    # ------------------------------------------------------------

    selected = sorted(
        valid_files
    )[:TARGET]

    # ------------------------------------------------------------
    # COPY 300
    # ------------------------------------------------------------

    copied = 0

    for i, src in enumerate(
        selected,
        start=1
    ):

        extension = os.path.splitext(
            src
        )[1].lower()

        destination_file = os.path.join(
            destination,
            f"{cls}_{i:03d}{extension}"
        )

        # Skip if already exists
        if os.path.exists(destination_file):

            copied += 1

            continue

        shutil.copy2(
            src,
            destination_file
        )

        copied += 1

    print(
        f"\n✅ {cls}"
    )

    print(
        f"   Available unique valid : {len(valid_files)}"
    )

    print(
        f"   Selected                : {TARGET}"
    )

    print(
        f"   Saved                   : {copied}"
    )

# ================================================================
# 11. FINAL VERIFIED_300 CHECK
# ================================================================

print("\n" + "=" * 75)
print("🍅 VERIFIED_300 FINAL CHECK")
print("=" * 75)

final_ok = True

for cls in CLASSES:

    folder = os.path.join(
        VERIFIED,
        cls
    )

    if not os.path.exists(folder):

        print(
            f"❌ {cls}: folder missing"
        )

        final_ok = False

        continue

    files_in_folder = [

        f for f in os.listdir(folder)

        if f.lower().endswith(
            IMAGE_EXTENSIONS
        )
    ]

    valid = 0
    corrupt = 0

    for filename in files_in_folder:

        path = os.path.join(
            folder,
            filename
        )

        try:

            with Image.open(path) as img:

                img.verify()

            valid += 1

        except:

            corrupt += 1

    print(
        f"{cls:15s} | "
        f"Files: {len(files_in_folder):3d} | "
        f"Valid: {valid:3d} | "
        f"Corrupt: {corrupt:2d}"
    )

    if (
        len(files_in_folder) != TARGET
        or valid != TARGET
        or corrupt != 0
    ):

        final_ok = False

# ================================================================
# 12. FINAL STATUS
# ================================================================

print("\n" + "=" * 75)

if final_ok:

    print("🎉 TOMATO EXTERNAL DATA VERIFIED SUCCESSFULLY")
    print("=" * 75)

    print("""
SOURCE:
Google Drive → 1st Raw_Data → Tomato → External_Natural

CLASSES:
Healthy       : 300
Early_Blight  : 300
Late_Blight   : 300

TOTAL:
900 verified images

QUALITY CHECK:
✅ Image validation
✅ Corrupt image detection
✅ Duplicate detection
✅ Format detection
✅ Dimension detection
✅ Exact 300/class selection

VERIFIED DATA:
External_Natural
└── Verified_300
    ├── Healthy
    ├── Early_Blight
    └── Late_Blight

REPORTS:
External_Natural
└── Verification_Report
    ├── Healthy_verification.csv
    ├── Early_Blight_verification.csv
    └── Late_Blight_verification.csv

NEXT:
➡️ Final Internal + External Verification
➡️ Then Data Cleaning / Preprocessing
""")

else:

    print("❌ VERIFICATION / SELECTION FAILED")
    print("=" * 75)

    print("""
Do NOT start preprocessing yet.

Check the class-wise results above.
""")

print("=" * 75)
```

```python
# ================================================================
# 🍅 TOMATO — FINAL INTERNAL + EXTERNAL VERIFICATION
# READ-ONLY AUDIT
# ================================================================

import os
import csv
import hashlib
from collections import Counter
from PIL import Image
from google.colab import drive

# ================================================================
# 1. GOOGLE DRIVE
# ================================================================

drive.mount("/content/drive")

# ================================================================
# 2. MAIN PATH
# ================================================================

PROJECT = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"

TOMATO = os.path.join(
    PROJECT,
    "1st Raw_Data",
    "Tomato"
)

INTERNAL = os.path.join(
    TOMATO,
    "Internal_Natural"
)

EXTERNAL = os.path.join(
    TOMATO,
    "External_Natural",
    "Verified_300"
)

REPORT_DIR = os.path.join(
    TOMATO,
    "Final_Verification"
)

os.makedirs(REPORT_DIR, exist_ok=True)

# ================================================================
# 3. EXPECTED CLASSES
# ================================================================

CLASSES = [
    "Healthy",
    "Early_Blight",
    "Late_Blight"
]

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp"
)

# ================================================================
# 4. HEADER
# ================================================================

print("=" * 75)
print("🍅 TOMATO — FINAL INTERNAL + EXTERNAL VERIFICATION")
print("=" * 75)

print("\nThis is a READ-ONLY verification.")
print("❌ No images will be deleted.")
print("❌ No images will be moved.")
print("❌ No images will be modified.")
print("✅ Only scanning and reporting will be performed.")

# ================================================================
# 5. CHECK MAIN FOLDERS
# ================================================================

print("\n" + "=" * 75)
print("📁 CHECKING DATA LOCATIONS")
print("=" * 75)

print("\nTomato folder:")
print(TOMATO)

if not os.path.exists(TOMATO):
    raise Exception(
        "❌ Tomato folder not found."
    )

print("✅ Tomato folder found.")

print("\nInternal:")
print(INTERNAL)

if os.path.exists(INTERNAL):
    print("✅ Internal_Natural found.")
else:
    print("⚠️ Internal_Natural NOT found.")

print("\nExternal:")
print(EXTERNAL)

if os.path.exists(EXTERNAL):
    print("✅ External Verified_300 found.")
else:
    raise Exception(
        "❌ External Verified_300 folder not found."
    )

# ================================================================
# 6. FUNCTION — VERIFY ONE IMAGE
# ================================================================

def verify_image(path):

    try:

        with Image.open(path) as img:

            image_format = img.format
            width, height = img.size

            # Force actual image verification
            img.verify()

        return {
            "valid": True,
            "format": image_format,
            "width": width,
            "height": height,
            "error": ""
        }

    except Exception as e:

        return {
            "valid": False,
            "format": "",
            "width": "",
            "height": "",
            "error": str(e)
        }


# ================================================================
# 7. FUNCTION — SCAN DATASET
# ================================================================

def scan_dataset(source_name, base_path):

    results = []

    print("\n" + "=" * 75)
    print(f"🔍 SCANNING: {source_name}")
    print("=" * 75)

    if not os.path.exists(base_path):

        print(
            f"⚠️ {source_name} folder does not exist."
        )

        return results

    for cls in CLASSES:

        class_path = os.path.join(
            base_path,
            cls
        )

        print("\n" + "-" * 75)
        print(f"CLASS: {cls}")
        print(f"PATH : {class_path}")
        print("-" * 75)

        if not os.path.exists(class_path):

            print("❌ CLASS FOLDER NOT FOUND")

            continue

        image_files = []

        for root, _, files in os.walk(class_path):

            for file in files:

                if file.lower().endswith(
                    IMAGE_EXTENSIONS
                ):

                    image_files.append(
                        os.path.join(root, file)
                    )

        print(
            "Image files found:",
            len(image_files)
        )

        valid_count = 0
        corrupt_count = 0

        hashes = set()

        duplicate_count = 0

        format_counter = Counter()

        dimension_counter = Counter()

        for path in image_files:

            check = verify_image(path)

            filename = os.path.basename(path)

            if check["valid"]:

                valid_count += 1

                format_counter[
                    check["format"]
                ] += 1

                dimension_counter[
                    f"{check['width']}x{check['height']}"
                ] += 1

                # ----------------------------------------
                # FILE HASH
                # ----------------------------------------

                try:

                    with open(
                        path,
                        "rb"
                    ) as f:

                        file_hash = hashlib.md5(
                            f.read()
                        ).hexdigest()

                    if file_hash in hashes:

                        duplicate_count += 1

                    else:

                        hashes.add(file_hash)

                except Exception:

                    pass

            else:

                corrupt_count += 1

            results.append({

                "Source": source_name,
                "Class": cls,
                "Filename": filename,
                "Path": path,
                "Valid": check["valid"],
                "Format": check["format"],
                "Width": check["width"],
                "Height": check["height"],
                "Error": check["error"]

            })

        print(
            f"Total    : {len(image_files)}"
        )

        print(
            f"Valid    : {valid_count}"
        )

        print(
            f"Corrupt  : {corrupt_count}"
        )

        print(
            f"Duplicate: {duplicate_count}"
        )

        print(
            f"Unique   : {len(hashes)}"
        )

        print("\nFormats:")

        for fmt, count in format_counter.items():

            print(
                f"  {fmt:10s}: {count}"
            )

        print("\nDimensions:")

        for dim, count in dimension_counter.items():

            print(
                f"  {dim:12s}: {count}"
            )

    return results


# ================================================================
# 8. SCAN INTERNAL
# ================================================================

internal_results = scan_dataset(
    "INTERNAL",
    INTERNAL
)

# ================================================================
# 9. SCAN EXTERNAL
# ================================================================

external_results = scan_dataset(
    "EXTERNAL",
    EXTERNAL
)

# ================================================================
# 10. COMBINE RESULTS
# ================================================================

all_results = (
    internal_results +
    external_results
)

# ================================================================
# 11. SAVE DETAILED CSV
# ================================================================

detail_csv = os.path.join(
    REPORT_DIR,
    "Tomato_Final_Detailed_Verification.csv"
)

with open(
    detail_csv,
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f,
        fieldnames=[
            "Source",
            "Class",
            "Filename",
            "Path",
            "Valid",
            "Format",
            "Width",
            "Height",
            "Error"
        ]
    )

    writer.writeheader()

    writer.writerows(all_results)

print("\n📄 Detailed report saved:")
print(detail_csv)

# ================================================================
# 12. CLASS SUMMARY
# ================================================================

print("\n" + "=" * 75)
print("📊 FINAL CLASS SUMMARY")
print("=" * 75)

summary_rows = []

for source in ["INTERNAL", "EXTERNAL"]:

    for cls in CLASSES:

        rows = [

            r for r in all_results

            if r["Source"] == source
            and r["Class"] == cls

        ]

        total = len(rows)

        valid = sum(
            1 for r in rows
            if r["Valid"]
        )

        corrupt = total - valid

        hashes = set()

        duplicates = 0

        for r in rows:

            if not r["Valid"]:
                continue

            try:

                with open(
                    r["Path"],
                    "rb"
                ) as f:

                    h = hashlib.md5(
                        f.read()
                    ).hexdigest()

                if h in hashes:

                    duplicates += 1

                else:

                    hashes.add(h)

            except:

                pass

        unique = len(hashes)

        formats = Counter(
            r["Format"]
            for r in rows
            if r["Valid"]
        )

        dimensions = Counter(
            f"{r['Width']}x{r['Height']}"
            for r in rows
            if r["Valid"]
        )

        format_text = ", ".join(
            f"{k}:{v}"
            for k, v in formats.items()
        )

        dimension_text = ", ".join(
            f"{k}:{v}"
            for k, v in dimensions.items()
        )

        summary_rows.append({

            "Source": source,
            "Class": cls,
            "Total": total,
            "Valid": valid,
            "Corrupt": corrupt,
            "Duplicates": duplicates,
            "Unique": unique,
            "Formats": format_text,
            "Dimensions": dimension_text

        })

        print(
            f"{source:8s} | "
            f"{cls:15s} | "
            f"Total: {total:3d} | "
            f"Valid: {valid:3d} | "
            f"Corrupt: {corrupt:2d} | "
            f"Duplicate: {duplicates:2d} | "
            f"Unique: {unique:3d}"
        )

# ================================================================
# 13. SAVE SUMMARY CSV
# ================================================================

summary_csv = os.path.join(
    REPORT_DIR,
    "Tomato_Final_Summary.csv"
)

with open(
    summary_csv,
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f,
        fieldnames=[
            "Source",
            "Class",
            "Total",
            "Valid",
            "Corrupt",
            "Duplicates",
            "Unique",
            "Formats",
            "Dimensions"
        ]
    )

    writer.writeheader()

    writer.writerows(summary_rows)

print("\n📄 Summary report saved:")
print(summary_csv)

# ================================================================
# 14. CROSS-SOURCE DUPLICATE CHECK
# ================================================================

print("\n" + "=" * 75)
print("🔗 INTERNAL ↔ EXTERNAL DUPLICATE CHECK")
print("=" * 75)

internal_hashes = {}
external_hashes = {}

for row in internal_results:

    if not row["Valid"]:
        continue

    try:

        with open(
            row["Path"],
            "rb"
        ) as f:

            h = hashlib.md5(
                f.read()
            ).hexdigest()

        internal_hashes.setdefault(
            h,
            []
        ).append(row["Path"])

    except:

        pass


for row in external_results:

    if not row["Valid"]:
        continue

    try:

        with open(
            row["Path"],
            "rb"
        ) as f:

            h = hashlib.md5(
                f.read()
            ).hexdigest()

        external_hashes.setdefault(
            h,
            []
        ).append(row["Path"])

    except:

        pass


common_hashes = set(
    internal_hashes.keys()
).intersection(
    external_hashes.keys()
)

print(
    "Internal unique hashes:",
    len(internal_hashes)
)

print(
    "External unique hashes:",
    len(external_hashes)
)

print(
    "Cross-source duplicate files:",
    len(common_hashes)
)

if len(common_hashes) == 0:

    print(
        "✅ No exact duplicate found between Internal and External."
    )

else:

    print(
        "⚠️ Exact duplicate content exists between Internal and External."
    )

# ================================================================
# 15. FINAL PASS / FAIL
# ================================================================

print("\n" + "=" * 75)
print("🏁 TOMATO FINAL VERIFICATION RESULT")
print("=" * 75)

final_pass = True

# ------------------------------------------------
# External MUST be exactly 300/class
# ------------------------------------------------

for row in summary_rows:

    source = row["Source"]
    cls = row["Class"]

    if source == "EXTERNAL":

        if (
            row["Total"] != 300
            or row["Valid"] != 300
            or row["Corrupt"] != 0
            or row["Duplicates"] != 0
        ):

            final_pass = False

            print(
                f"❌ EXTERNAL FAIL → {cls}"
            )

        else:

            print(
                f"✅ EXTERNAL PASS → {cls}"
            )

# ------------------------------------------------
# Internal quality check
# ------------------------------------------------

for row in summary_rows:

    source = row["Source"]
    cls = row["Class"]

    if source == "INTERNAL":

        if (
            row["Total"] == 0
            or row["Valid"] != row["Total"]
            or row["Corrupt"] != 0
            or row["Duplicates"] != 0
        ):

            final_pass = False

            print(
                f"❌ INTERNAL FAIL → {cls}"
            )

        else:

            print(
                f"✅ INTERNAL PASS → {cls}"
            )

# ------------------------------------------------
# Cross-source duplicate rule
# ------------------------------------------------

if len(common_hashes) != 0:

    final_pass = False

# ================================================================
# 16. FINAL STATUS
# ================================================================

print("\n" + "=" * 75)

if final_pass:

    print("🎉 TOMATO FINAL VERIFICATION PASSED")
    print("=" * 75)

    print("""
✅ Internal data verified
✅ External data verified
✅ Valid image check completed
✅ Corrupt image check completed
✅ Duplicate check completed
✅ Format check completed
✅ Dimension check completed
✅ Internal ↔ External duplicate check completed

🍅 TOMATO DATA IS READY FOR PREPROCESSING
""")

else:

    print("❌ TOMATO FINAL VERIFICATION FAILED")
    print("=" * 75)

    print("""
Do NOT start preprocessing yet.

Check the summary above and the CSV reports.
""")

print("=" * 75)

print("\n📁 Reports:")
print(REPORT_DIR)

print("\n➡️ NEXT STEP:")
print("3rd Preprocessing")
```

```python
# ================================================================
# 🍅 TOMATO — FINAL INTERNAL + EXTERNAL DATA VERIFICATION
# AUTOMATIC INTERNAL DATASET DISCOVERY
# READ-ONLY — NO DATA MODIFICATION
# ================================================================

import os
import csv
import hashlib
from collections import Counter
from PIL import Image
from google.colab import drive

# ================================================================
# 1. GOOGLE DRIVE
# ================================================================

drive.mount("/content/drive")

# ================================================================
# 2. MAIN PATHS
# ================================================================

PROJECT = "/content/drive/MyDrive/Plant Disease Detection (Computer Vision)"

TOMATO = os.path.join(
    PROJECT,
    "1st Raw_Data",
    "Tomato"
)

EXTERNAL = os.path.join(
    TOMATO,
    "External_Natural",
    "Verified_300"
)

REPORT_DIR = os.path.join(
    TOMATO,
    "Final_Verification"
)

os.makedirs(REPORT_DIR, exist_ok=True)

# ================================================================
# 3. REQUIRED CLASSES
# ================================================================

CLASSES = [
    "Healthy",
    "Early_Blight",
    "Late_Blight"
]

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp"
)

# ================================================================
# 4. HEADER
# ================================================================

print("=" * 75)
print("🍅 TOMATO — FINAL INTERNAL + EXTERNAL VERIFICATION")
print("=" * 75)

print("""
READ-ONLY VERIFICATION

✅ Scan
✅ Count
✅ Validate
✅ Duplicate check
✅ Format check
✅ Dimension check
✅ Internal ↔ External comparison

❌ No deletion
❌ No movement
❌ No renaming
❌ No resizing
❌ No preprocessing
""")

# ================================================================
# 5. CHECK TOMATO FOLDER
# ================================================================

if not os.path.exists(TOMATO):

    raise Exception(
        "❌ Tomato folder not found:\n" + TOMATO
    )

print("Tomato folder:")
print(TOMATO)
print("✅ Tomato folder found.")

# ================================================================
# 6. CHECK EXTERNAL VERIFIED_300
# ================================================================

print("\n" + "=" * 75)
print("📦 CHECKING EXTERNAL DATA")
print("=" * 75)

print(EXTERNAL)

if not os.path.exists(EXTERNAL):

    raise Exception(
        "❌ External Verified_300 not found."
    )

print("✅ External Verified_300 found.")

# ================================================================
# 7. FIND INTERNAL DATASET AUTOMATICALLY
# ================================================================

print("\n" + "=" * 75)
print("🔎 FINDING INTERNAL TOMATO DATASET")
print("=" * 75)

internal_candidates = []

for root, dirs, files in os.walk(TOMATO):

    # Ignore known external/report folders
    normalized = root.replace("\\", "/")

    if "/External_Natural" in normalized:
        continue

    if "/Final_Verification" in normalized:
        continue

    # Check whether this folder contains class folders
    folder_names = set(dirs)

    has_healthy = (
        "Healthy" in folder_names
        or "Healthy Leaf" in folder_names
    )

    has_early = (
        "Early_Blight" in folder_names
        or "Early Blight" in folder_names
        or "Early Blight Leaf" in folder_names
    )

    has_late = (
        "Late_Blight" in folder_names
        or "Late Blight" in folder_names
        or "Late Blight Leaf" in folder_names
    )

    if has_healthy and has_early and has_late:

        internal_candidates.append(root)

# ================================================================
# 8. SHOW CANDIDATES
# ================================================================

print("\nPossible Internal dataset locations:")

if len(internal_candidates) == 0:

    print("❌ No Internal dataset automatically found.")

    print("""
Please check your Google Drive:

1st Raw_Data
└── Tomato

The Internal dataset must contain:
    Healthy
    Early_Blight / Early Blight
    Late_Blight / Late Blight
""")

    raise Exception(
        "❌ Internal dataset location could not be identified."
    )

for i, path in enumerate(
    internal_candidates,
    start=1
):

    print(
        f"{i}. {path}"
    )

# ================================================================
# 9. SELECT INTERNAL DATASET
# ================================================================

if len(internal_candidates) > 1:

    print("""
⚠️ Multiple possible Internal datasets found.

The first matching location will be used for this verification.
""")

INTERNAL = internal_candidates[0]

print("\n✅ Internal dataset selected:")
print(INTERNAL)

# ================================================================
# 10. CLASS FOLDER FINDER
# ================================================================

def find_class_folder(base_path, class_name):

    aliases = {

        "Healthy": [
            "Healthy",
            "Healthy Leaf",
            "Healthy_Leaf"
        ],

        "Early_Blight": [
            "Early_Blight",
            "Early Blight",
            "Early Blight Leaf",
            "Early_Blight_Leaf"
        ],

        "Late_Blight": [
            "Late_Blight",
            "Late Blight",
            "Late Blight Leaf",
            "Late_Blight_Leaf"
        ]

    }

    for alias in aliases[class_name]:

        path = os.path.join(
            base_path,
            alias
        )

        if os.path.isdir(path):

            return path

    return None

# ================================================================
# 11. IMAGE VERIFICATION
# ================================================================

def verify_image(path):

    try:

        with Image.open(path) as img:

            image_format = img.format
            width, height = img.size

            img.verify()

        return {
            "valid": True,
            "format": image_format,
            "width": width,
            "height": height,
            "error": ""
        }

    except Exception as e:

        return {
            "valid": False,
            "format": "",
            "width": "",
            "height": "",
            "error": str(e)
        }

# ================================================================
# 12. SCAN DATASET
# ================================================================

def scan_dataset(source_name, base_path):

    results = []

    print("\n" + "=" * 75)
    print("🔍 SCANNING:", source_name)
    print("=" * 75)

    for cls in CLASSES:

        class_path = find_class_folder(
            base_path,
            cls
        )

        print("\n" + "-" * 75)
        print("CLASS:", cls)

        if class_path is None:

            print("❌ Class folder NOT FOUND")

            continue

        print("PATH:", class_path)

        image_files = []

        for root, _, files in os.walk(class_path):

            for file in files:

                if file.lower().endswith(
                    IMAGE_EXTENSIONS
                ):

                    image_files.append(
                        os.path.join(
                            root,
                            file
                        )
                    )

        print(
            "Image files found:",
            len(image_files)
        )

        valid = 0
        corrupt = 0
        hashes = set()
        duplicate = 0

        formats = Counter()
        dimensions = Counter()

        for path in image_files:

            result = verify_image(path)

            if result["valid"]:

                valid += 1

                formats[
                    result["format"]
                ] += 1

                dimensions[
                    f"{result['width']}x{result['height']}"
                ] += 1

                try:

                    with open(
                        path,
                        "rb"
                    ) as f:

                        h = hashlib.md5(
                            f.read()
                        ).hexdigest()

                    if h in hashes:

                        duplicate += 1

                    else:

                        hashes.add(h)

                except Exception:

                    pass

            else:

                corrupt += 1

            results.append({

                "Source": source_name,
                "Class": cls,
                "Filename": os.path.basename(path),
                "Path": path,
                "Valid": result["valid"],
                "Format": result["format"],
                "Width": result["width"],
                "Height": result["height"],
                "Error": result["error"]

            })

        print(
            f"Total     : {len(image_files)}"
        )

        print(
            f"Valid     : {valid}"
        )

        print(
            f"Corrupt   : {corrupt}"
        )

        print(
            f"Duplicates: {duplicate}"
        )

        print(
            f"Unique    : {len(hashes)}"
        )

        print("\nFormats:")

        for key, value in formats.items():

            print(
                f"  {key}: {value}"
            )

        print("\nDimensions:")

        for key, value in dimensions.items():

            print(
                f"  {key}: {value}"
            )

    return results

# ================================================================
# 13. SCAN INTERNAL
# ================================================================

internal_results = scan_dataset(
    "INTERNAL",
    INTERNAL
)

# ================================================================
# 14. SCAN EXTERNAL
# ================================================================

external_results = scan_dataset(
    "EXTERNAL",
    EXTERNAL
)

# ================================================================
# 15. COMBINE
# ================================================================

all_results = (
    internal_results +
    external_results
)

# ================================================================
# 16. CROSS SOURCE DUPLICATE CHECK
# ================================================================

print("\n" + "=" * 75)
print("🔗 INTERNAL ↔ EXTERNAL DUPLICATE CHECK")
print("=" * 75)

internal_hashes = set()
external_hashes = set()

for row in internal_results:

    if not row["Valid"]:
        continue

    try:

        with open(
            row["Path"],
            "rb"
        ) as f:

            h = hashlib.md5(
                f.read()
            ).hexdigest()

        internal_hashes.add(h)

    except:

        pass


for row in external_results:

    if not row["Valid"]:
        continue

    try:

        with open(
            row["Path"],
            "rb"
        ) as f:

            h = hashlib.md5(
                f.read()
            ).hexdigest()

        external_hashes.add(h)

    except:

        pass


common = (
    internal_hashes
    &
    external_hashes
)

print(
    "Internal unique images:",
    len(internal_hashes)
)

print(
    "External unique images:",
    len(external_hashes)
)

print(
    "Cross-source exact duplicates:",
    len(common)
)

if len(common) == 0:

    print(
        "✅ No exact Internal ↔ External duplicates."
    )

else:

    print(
        "⚠️ Cross-source duplicates detected."
    )

# ================================================================
# 17. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 75)
print("📊 FINAL TOMATO SUMMARY")
print("=" * 75)

summary = []

for source in [
    "INTERNAL",
    "EXTERNAL"
]:

    for cls in CLASSES:

        rows = [

            r for r in all_results

            if r["Source"] == source
            and r["Class"] == cls

        ]

        total = len(rows)

        valid = sum(
            r["Valid"]
            for r in rows
        )

        corrupt = total - valid

        hashes = set()
        duplicates = 0

        for r in rows:

            if not r["Valid"]:
                continue

            try:

                with open(
                    r["Path"],
                    "rb"
                ) as f:

                    h = hashlib.md5(
                        f.read()
                    ).hexdigest()

                if h in hashes:

                    duplicates += 1

                else:

                    hashes.add(h)

            except:

                pass

        unique = len(hashes)

        summary.append({

            "Source": source,
            "Class": cls,
            "Total": total,
            "Valid": valid,
            "Corrupt": corrupt,
            "Duplicates": duplicates,
            "Unique": unique

        })

        print(
            f"{source:8s} | "
            f"{cls:15s} | "
            f"Total: {total:3d} | "
            f"Valid: {valid:3d} | "
            f"Corrupt: {corrupt:2d} | "
            f"Duplicate: {duplicates:2d} | "
            f"Unique: {unique:3d}"
        )

# ================================================================
# 18. SAVE REPORTS
# ================================================================

detail_report = os.path.join(
    REPORT_DIR,
    "Tomato_Final_Detailed_Verification.csv"
)

with open(
    detail_report,
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f,
        fieldnames=[
            "Source",
            "Class",
            "Filename",
            "Path",
            "Valid",
            "Format",
            "Width",
            "Height",
            "Error"
        ]
    )

    writer.writeheader()

    writer.writerows(
        all_results
    )


summary_report = os.path.join(
    REPORT_DIR,
    "Tomato_Final_Summary.csv"
)

with open(
    summary_report,
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f,
        fieldnames=[
            "Source",
            "Class",
            "Total",
            "Valid",
            "Corrupt",
            "Duplicates",
            "Unique"
        ]
    )

    writer.writeheader()

    writer.writerows(
        summary
    )

print("\n📄 Reports saved:")

print(detail_report)
print(summary_report)

# ================================================================
# 19. FINAL PASS / FAIL
# ================================================================

print("\n" + "=" * 75)
print("🏁 FINAL TOMATO VERIFICATION")
print("=" * 75)

FINAL_PASS = True

# External must be exactly 300/class
for row in summary:

    if row["Source"] == "EXTERNAL":

        if (
            row["Total"] != 300
            or row["Valid"] != 300
            or row["Corrupt"] != 0
            or row["Duplicates"] != 0
        ):

            FINAL_PASS = False

            print(
                f"❌ EXTERNAL FAIL → {row['Class']}"
            )

        else:

            print(
                f"✅ EXTERNAL PASS → {row['Class']}"
            )

# Internal must exist and be clean
for row in summary:

    if row["Source"] == "INTERNAL":

        if (
            row["Total"] == 0
            or row["Valid"] != row["Total"]
            or row["Corrupt"] != 0
            or row["Duplicates"] != 0
        ):

            FINAL_PASS = False

            print(
                f"❌ INTERNAL FAIL → {row['Class']}"
            )

        else:

            print(
                f"✅ INTERNAL PASS → {row['Class']}"
            )

# Cross-source duplicate rule
if len(common) != 0:

    FINAL_PASS = False

# ================================================================
# 20. FINAL RESULT
# ================================================================

print("\n" + "=" * 75)

if FINAL_PASS:

    print("🎉 TOMATO — FINAL VERIFICATION PASSED")
    print("=" * 75)

    print("""
✅ Internal dataset verified
✅ External dataset verified
✅ All required classes verified
✅ Valid image check passed
✅ Corrupt image check passed
✅ Duplicate check passed
✅ Format check completed
✅ Dimension check completed
✅ Internal ↔ External duplicate check passed

🍅 TOMATO DATA IS READY

NEXT:
➡️ Data Cleaning
➡️ Preprocessing
""")

else:

    print("❌ TOMATO — FINAL VERIFICATION FAILED")
    print("=" * 75)

    print("""
Do NOT start preprocessing yet.

Read the class-wise results above.

If Internal was detected under a different folder,
the detected path is printed above.
""")

print("=" * 75)
```

```python
# ================================================================
# 🍅 TOMATO — IDENTIFY INTERNAL HEALTHY DUPLICATE
# READ-ONLY
# ================================================================

import os
import hashlib

HEALTHY_PATH = (
    "/content/drive/MyDrive/"
    "Plant Disease Detection (Computer Vision)/"
    "1st Raw_Data/Tomato/"
    "Internal_PlantVillage/Healthy"
)

print("=" * 70)
print("🍅 FINDING EXACT DUPLICATE — INTERNAL HEALTHY")
print("=" * 70)

hash_map = {}
duplicates = []

for file in sorted(os.listdir(HEALTHY_PATH)):

    path = os.path.join(HEALTHY_PATH, file)

    if not os.path.isfile(path):
        continue

    if not file.lower().endswith(
        (".jpg", ".jpeg", ".png", ".webp")
    ):
        continue

    h = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)

    file_hash = h.hexdigest()

    if file_hash in hash_map:

        duplicates.append(
            (hash_map[file_hash], file)
        )

    else:

        hash_map[file_hash] = file


print("\nTotal images :", len(hash_map) + len(duplicates))
print("Unique       :", len(hash_map))
print("Duplicates   :", len(duplicates))

print("\n" + "=" * 70)
print("DUPLICATE FILES")
print("=" * 70)

for original, duplicate in duplicates:

    print("\nDuplicate pair:")
    print("1 →", original)
    print("2 →", duplicate)

print("\n" + "=" * 70)
print("NO FILES WERE DELETED OR MODIFIED.")
print("=" * 70)
```

```python
# ================================================================
# 🍅 TOMATO — FIND 1 REPLACEMENT HEALTHY IMAGE
# READ-ONLY — NOTHING DELETED OR MODIFIED
# ================================================================

import os
import hashlib
from PIL import Image

# ------------------------------------------------
# CURRENT HEALTHY FOLDER
# ------------------------------------------------

CURRENT = (
    "/content/drive/MyDrive/"
    "Plant Disease Detection (Computer Vision)/"
    "1st Raw_Data/Tomato/"
    "Internal_PlantVillage/Healthy"
)

# ------------------------------------------------
# ORIGINAL SOURCE ROOT
# ------------------------------------------------

SOURCE_ROOT = (
    "/content/drive/MyDrive/"
    "Plant Disease Detection (Computer Vision)/"
    "1st Raw_Data/Tomato"
)

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp"
)

print("=" * 70)
print("🍅 FINDING 1 VALID HEALTHY REPLACEMENT")
print("=" * 70)

# ------------------------------------------------
# 1. HASH CURRENT HEALTHY IMAGES
# ------------------------------------------------

current_hashes = set()

current_files = []

for file in os.listdir(CURRENT):

    path = os.path.join(CURRENT, file)

    if not os.path.isfile(path):
        continue

    if not file.lower().endswith(IMAGE_EXTENSIONS):
        continue

    try:

        with Image.open(path) as img:
            img.verify()

        h = hashlib.sha256()

        with open(path, "rb") as f:
            for chunk in iter(
                lambda: f.read(1024 * 1024),
                b""
            ):
                h.update(chunk)

        current_hashes.add(h.hexdigest())
        current_files.append(file)

    except:
        pass

print("Current Healthy files :", len(current_files))
print("Current unique hashes :", len(current_hashes))

# ------------------------------------------------
# 2. SEARCH FOR ADDITIONAL HEALTHY IMAGES
# ------------------------------------------------

print("\nSearching Tomato Internal source...")

candidates = []

for root, dirs, files in os.walk(SOURCE_ROOT):

    # Don't search the current Healthy output again
    if os.path.abspath(root) == os.path.abspath(CURRENT):
        continue

    # Don't search External
    if "External_Natural" in root:
        continue

    # Don't search verification reports
    if "Final_Verification" in root:
        continue

    # Only Healthy-related folders
    folder_text = root.lower().replace("_", " ")

    if "healthy" not in folder_text:
        continue

    for file in files:

        if not file.lower().endswith(IMAGE_EXTENSIONS):
            continue

        path = os.path.join(root, file)

        try:

            with Image.open(path) as img:
                img.verify()

            h = hashlib.sha256()

            with open(path, "rb") as f:
                for chunk in iter(
                    lambda: f.read(1024 * 1024),
                    b""
                ):
                    h.update(chunk)

            file_hash = h.hexdigest()

            if file_hash not in current_hashes:

                candidates.append(
                    (path, file_hash)
                )

        except:
            continue

# ------------------------------------------------
# 3. RESULT
# ------------------------------------------------

print("\n" + "=" * 70)
print("RESULT")
print("=" * 70)

print(
    "Additional valid unique Healthy images found:",
    len(candidates)
)

if len(candidates) > 0:

    print("\n✅ Replacement candidate(s) available:\n")

    for i, (path, h) in enumerate(
        candidates[:10],
        start=1
    ):

        print(f"{i}. {path}")

else:

    print("""
❌ No additional Healthy image was found.

The current Internal Healthy dataset contains
300 files but only 299 unique images.

We will need to obtain ONE additional original
PlantVillage Healthy image before final verification.
""")

print("\n" + "=" * 70)
print("READ-ONLY SCAN COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# 🍅 TOMATO — FINAL DATA LOCK
# INTERNAL + EXTERNAL
# ================================================================
# ACCEPTED RULE:
# Internal Healthy = 299 unique is acceptable
# All other classes = 300 unique required
#
# READ-ONLY
# ❌ No deletion
# ❌ No movement
# ❌ No renaming
# ❌ No modification
# ================================================================

import os
import hashlib
from PIL import Image
from google.colab import drive

drive.mount("/content/drive")

# ------------------------------------------------
# PATHS
# ------------------------------------------------

PROJECT = (
    "/content/drive/MyDrive/"
    "Plant Disease Detection (Computer Vision)"
)

TOMATO = os.path.join(
    PROJECT,
    "1st Raw_Data",
    "Tomato"
)

INTERNAL = os.path.join(
    TOMATO,
    "Internal_PlantVillage"
)

EXTERNAL = os.path.join(
    TOMATO,
    "External_Natural",
    "Verified_300"
)

CLASSES = [
    "Healthy",
    "Early_Blight",
    "Late_Blight"
]

# ------------------------------------------------
# SCANNER
# ------------------------------------------------

def scan_folder(folder):

    total = 0
    valid = 0
    corrupt = 0

    hashes = set()
    duplicates = 0

    formats = {}
    dimensions = {}

    if not os.path.exists(folder):
        return None

    for file in os.listdir(folder):

        path = os.path.join(folder, file)

        if not os.path.isfile(path):
            continue

        if not file.lower().endswith(
            (".jpg", ".jpeg", ".png", ".webp")
        ):
            continue

        total += 1

        try:

            with Image.open(path) as img:

                img.verify()

            with Image.open(path) as img:

                fmt = img.format
                size = img.size

            formats[fmt] = formats.get(fmt, 0) + 1

            dimensions[size] = (
                dimensions.get(size, 0) + 1
            )

            with open(path, "rb") as f:

                h = hashlib.sha256(
                    f.read()
                ).hexdigest()

            if h in hashes:

                duplicates += 1

            else:

                hashes.add(h)

            valid += 1

        except:

            corrupt += 1

    return {
        "total": total,
        "valid": valid,
        "corrupt": corrupt,
        "duplicates": duplicates,
        "unique": len(hashes),
        "formats": formats,
        "dimensions": dimensions
    }


# ================================================================
# FINAL SCAN
# ================================================================

print("=" * 72)
print("🍅 TOMATO — FINAL DATA LOCK VERIFICATION")
print("=" * 72)

results = {}

for source_name, base in [
    ("INTERNAL", INTERNAL),
    ("EXTERNAL", EXTERNAL)
]:

    print("\n" + "=" * 72)
    print(source_name)
    print("=" * 72)

    for cls in CLASSES:

        folder = os.path.join(
            base,
            cls
        )

        r = scan_folder(folder)

        results[
            (source_name, cls)
        ] = r

        if r is None:

            print(
                f"{cls:15s} | ❌ FOLDER NOT FOUND"
            )

            continue

        print(
            f"{cls:15s} | "
            f"Files: {r['total']:3d} | "
            f"Valid: {r['valid']:3d} | "
            f"Corrupt: {r['corrupt']:2d} | "
            f"Duplicate: {r['duplicates']:2d} | "
            f"Unique: {r['unique']:3d}"
        )

        print(
            f"                 "
            f"Formats: {r['formats']}"
        )

        print(
            f"                 "
            f"Dimensions: {r['dimensions']}"
        )


# ================================================================
# ACCEPTANCE RULE
# ================================================================

print("\n" + "=" * 72)
print("🍅 FINAL ACCEPTANCE CHECK")
print("=" * 72)

passed = True

for (source, cls), r in results.items():

    if r is None:
        passed = False
        continue

    # Internal Healthy → minimum 299 unique
    if source == "INTERNAL" and cls == "Healthy":

        if (
            r["unique"] >= 299
            and r["corrupt"] == 0
            and r["duplicates"] <= 1
        ):

            print(
                f"✅ {source} | {cls} → ACCEPTED "
                f"({r['unique']} unique)"
            )

        else:

            print(
                f"❌ {source} | {cls} → FAILED"
            )

            passed = False

    # Everything else → exactly 300 unique
    else:

        if (
            r["unique"] == 300
            and r["corrupt"] == 0
            and r["duplicates"] == 0
        ):

            print(
                f"✅ {source} | {cls} → PASS"
            )

        else:

            print(
                f"❌ {source} | {cls} → FAILED"
            )

            passed = False


# ================================================================
# FINAL RESULT
# ================================================================

print("\n" + "=" * 72)

if passed:

    print("🎉 TOMATO DATASET — FINAL PASS")
    print("=" * 72)

    print("""
Internal:
    Healthy       = 299 unique
    Early_Blight  = 300 unique
    Late_Blight   = 300 unique

External:
    Healthy       = 300 unique
    Early_Blight  = 300 unique
    Late_Blight   = 300 unique

TOTAL UNIQUE = 1,799

✅ Data accepted
✅ No corrupt images
✅ Duplicate issue accepted only for Internal Healthy
✅ Internal and External remain separate
✅ No re-download required

================================================
🍅 TOMATO DATA IS NOW LOCKED
================================================

NEXT:
➡️ DATA CLEANING + PREPROCESSING
""")

else:

    print("❌ TOMATO DATASET — FINAL CHECK FAILED")
    print("Do not start preprocessing.")

print("=" * 72)
```
