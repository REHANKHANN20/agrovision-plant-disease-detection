```python
# ================================================================
# 🍅 TOMATO CNN — FINAL MASTER MODEL EVALUATION
# SAFE / READ-ONLY MODEL + DATASET
# ================================================================

import os
import json
import csv
import hashlib
import random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_recall_fscore_support
)

print("=" * 80)
print("🍅 TOMATO CNN — FINAL MASTER MODEL EVALUATION")
print("=" * 80)

print("TensorFlow version :", tf.__version__)

SEED = 42
print("Random seed        :", SEED)

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# ================================================================
# 1. GOOGLE DRIVE — ROBUST MOUNT
# ================================================================

print("\n" + "=" * 80)
print("☁️ GOOGLE DRIVE CHECK")
print("=" * 80)

from google.colab import drive

DRIVE_MOUNT = "/content/drive"

# Mount only if needed
try:
    drive.mount(DRIVE_MOUNT, force_remount=False)
except Exception as e:
    print("⚠️ Normal mount attempt returned:", str(e))

print("\nChecking Drive structure...")

possible_drive_roots = [
    "/content/drive/MyDrive",
    "/content/drive/My Drive"
]

MYDRIVE = None

for candidate in possible_drive_roots:

    if os.path.isdir(candidate):

        MYDRIVE = candidate
        break

if MYDRIVE is None:

    print("\n❌ Could not locate MyDrive.")

    print("\nContents of /content/drive:")

    if os.path.isdir("/content/drive"):

        for item in os.listdir("/content/drive"):
            print("   •", item)

    raise RuntimeError(
        "\nGoogle Drive is mounted, "
        "but MyDrive could not be located."
    )

print("✅ Google Drive accessible.")
print("MyDrive path:", MYDRIVE)


# ================================================================
# 2. PROJECT PATH
# ================================================================

PROJECT = os.path.join(
    MYDRIVE,
    "Plant Disease Detection (Computer Vision)"
)

print("\n" + "=" * 80)
print("📁 PROJECT PATH CHECK")
print("=" * 80)

print("Project:")
print(PROJECT)

if not os.path.isdir(PROJECT):

    print("\n❌ Project folder not found.")

    print("\nVisible folders inside MyDrive:")

    for item in os.listdir(MYDRIVE):

        full = os.path.join(MYDRIVE, item)

        if os.path.isdir(full):
            print("   📁", item)

    raise FileNotFoundError(
        "\nPlant Disease Detection (Computer Vision) "
        "folder was not found."
    )

print("✅ Project folder found.")


# ================================================================
# 3. DEFINE EXPECTED PATHS
# ================================================================

PREPROCESSING_DIR = os.path.join(
    PROJECT,
    "3rd Preprocessing"
)

TRAINING_DIR = os.path.join(
    PROJECT,
    "4th Model_Training"
)

EVALUATION_DIR = os.path.join(
    PROJECT,
    "5th Model_Evaluation"
)

MODEL_DIR = os.path.join(
    PROJECT,
    "6th Trained_Model"
)

DATASET_FILENAME = "tomato_processed_data.npz"

MODEL_FILENAME = "tomato_cnn_best.keras"

DATASET_PATH = os.path.join(
    PREPROCESSING_DIR,
    DATASET_FILENAME
)

MODEL_PATH = os.path.join(
    MODEL_DIR,
    MODEL_FILENAME
)


print("\nDataset:")
print(DATASET_PATH)

print("\nModel:")
print(MODEL_PATH)

print("\nEvaluation folder:")
print(EVALUATION_DIR)


# ================================================================
# 4. ROBUST DATASET SEARCH
# ================================================================

print("\n" + "=" * 80)
print("🔎 LOCATING TOMATO DATASET")
print("=" * 80)

if not os.path.isfile(DATASET_PATH):

    print("⚠️ Expected dataset path is not visible.")
    print("Searching inside project folder...")

    matches = []

    for root, dirs, files in os.walk(PROJECT):

        for file in files:

            if file == DATASET_FILENAME:

                matches.append(
                    os.path.join(root, file)
                )

    if len(matches) == 1:

        DATASET_PATH = matches[0]

        print("\n✅ Dataset found at:")
        print(DATASET_PATH)

    elif len(matches) > 1:

        print("\n❌ Multiple Tomato datasets found:")

        for path in matches:
            print("   •", path)

        raise RuntimeError(
            "\nMultiple tomato_processed_data.npz files "
            "were found. Evaluation stopped to prevent "
            "using the wrong dataset."
        )

    else:

        print("\n❌ Tomato dataset could not be found.")

        if os.path.isdir(PREPROCESSING_DIR):

            print(
                "\nFiles visible in 3rd Preprocessing:"
            )

            for item in sorted(
                os.listdir(PREPROCESSING_DIR)
            ):
                print("   •", item)

        raise FileNotFoundError(
            "\nTomato dataset was not found anywhere "
            "inside the project folder."
        )

else:

    print("✅ Tomato dataset found at expected path.")


# ================================================================
# 5. MODEL SEARCH
# ================================================================

print("\n" + "=" * 80)
print("🧠 LOCATING FINAL TOMATO MODEL")
print("=" * 80)

if not os.path.isfile(MODEL_PATH):

    print("⚠️ Expected model path is not visible.")
    print("Searching inside project folder...")

    model_matches = []

    for root, dirs, files in os.walk(PROJECT):

        for file in files:

            if file == MODEL_FILENAME:

                model_matches.append(
                    os.path.join(root, file)
                )

    if len(model_matches) == 1:

        MODEL_PATH = model_matches[0]

        print("\n✅ Model found at:")
        print(MODEL_PATH)

    elif len(model_matches) > 1:

        print("\n❌ Multiple Tomato models found:")

        for path in model_matches:
            print("   •", path)

        raise RuntimeError(
            "\nMultiple tomato_cnn_best.keras files "
            "were found. Evaluation stopped."
        )

    else:

        raise FileNotFoundError(
            "\nFinal Tomato CNN model was not found."
        )

else:

    print("✅ Final Tomato CNN found.")


# ================================================================
# 6. EVALUATION DIRECTORY
# ================================================================

print("\n" + "=" * 80)
print("📁 EVALUATION DIRECTORY")
print("=" * 80)

os.makedirs(
    EVALUATION_DIR,
    exist_ok=True
)

print("✅ Evaluation folder ready.")


# ================================================================
# 7. LOAD DATASET
# ================================================================

print("\n" + "=" * 80)
print("📥 LOADING TOMATO DATASET")
print("=" * 80)

data = np.load(
    DATASET_PATH,
    allow_pickle=True
)

print("Stored arrays:")
print(data.files)

required_arrays = {
    "X",
    "y",
    "class_names",
    "source"
}

missing = required_arrays - set(data.files)

if missing:

    raise ValueError(
        f"Missing required arrays: {sorted(missing)}"
    )

X = data["X"]
y = data["y"]
class_names = data["class_names"]
source = data["source"]

print("\nX shape        :", X.shape)
print("X dtype        :", X.dtype)

print("y shape        :", y.shape)
print("y dtype        :", y.dtype)

print("class_names    :", class_names)

print("source shape   :", source.shape)


# ================================================================
# 8. DATASET VALIDATION
# ================================================================

print("\n" + "=" * 80)
print("🔎 DATASET VALIDATION")
print("=" * 80)

assert X.shape == (1800, 224, 224, 3)

assert y.shape == (1800,)

assert len(class_names) == 3

assert len(source) == 1800

assert X.dtype == np.uint8

assert y.dtype == np.int64

print("✅ 1,800 images verified")
print("✅ 224 × 224 × 3 verified")
print("✅ uint8 images verified")
print("✅ Three classes verified")
print("✅ Source information verified")


# ================================================================
# 9. CLASS DISTRIBUTION
# ================================================================

print("\n" + "=" * 80)
print("📊 CLASS DISTRIBUTION")
print("=" * 80)

for i, name in enumerate(class_names):

    count = int(np.sum(y == i))

    print(
        f"{i} → {name:<15} : {count}"
    )

    assert count == 600

print("\n✅ Dataset is perfectly balanced.")


# ================================================================
# 10. EXACT DUPLICATE DETECTION
# ================================================================

print("\n" + "=" * 80)
print("🔍 EXACT DUPLICATE CHECK")
print("=" * 80)

hash_to_indices = {}

for i in range(len(X)):

    digest = hashlib.sha256(
        X[i].tobytes()
    ).hexdigest()

    hash_to_indices.setdefault(
        digest,
        []
    ).append(i)

duplicate_groups = [
    indices
    for indices in hash_to_indices.values()
    if len(indices) > 1
]

print("Total images         :", len(X))
print("Unique image contents:", len(hash_to_indices))
print("Duplicate groups     :", len(duplicate_groups))

if duplicate_groups:

    print("\nDuplicate groups:")

    for group in duplicate_groups:

        print("   ", group)

else:

    print("✅ No exact duplicates found.")


# ================================================================
# 11. BUILD DUPLICATE-SAFE GROUPS
# ================================================================

print("\n" + "=" * 80)
print("🔐 BUILDING DUPLICATE-SAFE GROUPS")
print("=" * 80)

groups = list(hash_to_indices.values())

print("Total image groups:", len(groups))


# ================================================================
# 12. CREATE DETERMINISTIC BALANCED SPLIT
# ================================================================

print("\n" + "=" * 80)
print("📊 RECREATING TRAIN / VALIDATION / TEST SPLIT")
print("=" * 80)

# ------------------------------------------------
# IMPORTANT:
# Each duplicate group stays together.
# Classes remain balanced.
# ------------------------------------------------

rng = np.random.RandomState(SEED)

class_groups = {
    i: []
    for i in range(len(class_names))
}

for group in groups:

    labels = set(
        int(y[idx])
        for idx in group
    )

    if len(labels) != 1:

        raise ValueError(
            f"Duplicate group contains multiple classes: "
            f"{group}"
        )

    label = next(iter(labels))

    class_groups[label].append(group)


# Shuffle groups deterministically
for label in class_groups:

    rng.shuffle(
        class_groups[label]
    )


# Desired counts
TRAIN_PER_CLASS = 420
VAL_PER_CLASS = 90
TEST_PER_CLASS = 90


train_indices = []
val_indices = []
test_indices = []


# ------------------------------------------------
# Select groups until desired image counts reached
# ------------------------------------------------

for label in range(len(class_names)):

    groups_for_class = class_groups[label]

    train_groups = []
    remaining = list(groups_for_class)

    # Select groups for TRAIN
    total = 0

    while remaining and total < TRAIN_PER_CLASS:

        group = remaining.pop()

        train_groups.append(group)

        total += len(group)

    if total != TRAIN_PER_CLASS:

        raise RuntimeError(
            f"Could not create exactly "
            f"{TRAIN_PER_CLASS} training images "
            f"for class {class_names[label]}."
        )

    # Select groups for VALIDATION
    val_groups = []
    total = 0

    while remaining and total < VAL_PER_CLASS:

        group = remaining.pop()

        val_groups.append(group)

        total += len(group)

    if total != VAL_PER_CLASS:

        raise RuntimeError(
            f"Could not create exactly "
            f"{VAL_PER_CLASS} validation images "
            f"for class {class_names[label]}."
        )

    # Everything remaining becomes TEST
    test_groups = remaining

    test_count = sum(
        len(g)
        for g in test_groups
    )

    if test_count != TEST_PER_CLASS:

        raise RuntimeError(
            f"Could not create exactly "
            f"{TEST_PER_CLASS} test images "
            f"for class {class_names[label]}."
        )

    for group in train_groups:
        train_indices.extend(group)

    for group in val_groups:
        val_indices.extend(group)

    for group in test_groups:
        test_indices.extend(group)


# Convert to numpy arrays
train_indices = np.array(
    sorted(train_indices),
    dtype=np.int64
)

val_indices = np.array(
    sorted(val_indices),
    dtype=np.int64
)

test_indices = np.array(
    sorted(test_indices),
    dtype=np.int64
)


# ================================================================
# 13. SPLIT VERIFICATION
# ================================================================

print("\n" + "=" * 80)
print("🔐 VERIFYING RECREATED SPLIT")
print("=" * 80)

print("Train      :", len(train_indices))
print("Validation :", len(val_indices))
print("Test       :", len(test_indices))

assert len(train_indices) == 1260

assert len(val_indices) == 270

assert len(test_indices) == 270

all_indices = np.concatenate([
    train_indices,
    val_indices,
    test_indices
])

assert len(np.unique(all_indices)) == 1800

assert set(train_indices).isdisjoint(
    set(val_indices)
)

assert set(train_indices).isdisjoint(
    set(test_indices)
)

assert set(val_indices).isdisjoint(
    set(test_indices)
)

print("\n✅ All 1,800 images accounted for.")
print("✅ No train/validation overlap.")
print("✅ No train/test overlap.")
print("✅ No validation/test overlap.")


# ================================================================
# 14. TEST CLASS DISTRIBUTION
# ================================================================

print("\n" + "=" * 80)
print("📊 TEST SET CLASS DISTRIBUTION")
print("=" * 80)

y_test = y[test_indices]

for i, name in enumerate(class_names):

    count = int(
        np.sum(y_test == i)
    )

    print(
        f"{name:<15} : {count}"
    )

    assert count == 90

print("\n✅ Test set is perfectly balanced: 90 / 90 / 90.")


# ================================================================
# 15. LOAD FINAL MODEL
# ================================================================

print("\n" + "=" * 80)
print("🧠 LOADING FINAL TOMATO CNN")
print("=" * 80)

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

print("✅ Model loaded successfully.")

print("Input shape :", model.input_shape)
print("Output shape:", model.output_shape)
print("Parameters  :", model.count_params())

assert model.input_shape == (
    None,
    224,
    224,
    3
)

assert model.output_shape == (
    None,
    3
)

print("✅ Model compatibility verified.")


# ================================================================
# 16. PREPARE TEST DATA
# ================================================================

print("\n" + "=" * 80)
print("🧪 PREPARING TEST DATA")
print("=" * 80)

X_test = X[test_indices]

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

assert X_test.shape == (
    270,
    224,
    224,
    3
)

assert y_test.shape == (
    270,
)

print("✅ Test data ready.")


# ================================================================
# 17. FINAL MODEL PREDICTION
# ================================================================

print("\n" + "=" * 80)
print("🚀 RUNNING FINAL TOMATO CNN EVALUATION")
print("=" * 80)

predictions = model.predict(
    X_test,
    batch_size=32,
    verbose=1
)

y_pred = np.argmax(
    predictions,
    axis=1
)

confidence = np.max(
    predictions,
    axis=1
)

print("\n✅ Predictions completed.")

assert predictions.shape == (
    270,
    3
)

assert y_pred.shape == (
    270,
)


# ================================================================
# 18. METRICS
# ================================================================

print("\n" + "=" * 80)
print("📈 FINAL TEST METRICS")
print("=" * 80)

test_accuracy = accuracy_score(
    y_test,
    y_pred
)

precision, recall, f1, support = (
    precision_recall_fscore_support(
        y_test,
        y_pred,
        labels=list(range(3)),
        zero_division=0
    )
)

print(
    f"\nTest Accuracy : {test_accuracy:.4f}"
)

print(
    f"Test Accuracy : {test_accuracy * 100:.2f}%"
)

print("\nPer-class metrics:")

for i, name in enumerate(class_names):

    print(
        f"{name:<15} | "
        f"Precision: {precision[i]:.4f} | "
        f"Recall: {recall[i]:.4f} | "
        f"F1: {f1[i]:.4f} | "
        f"Support: {support[i]}"
    )


# ================================================================
# 19. CONFUSION MATRIX
# ================================================================

print("\n" + "=" * 80)
print("🔲 CONFUSION MATRIX")
print("=" * 80)

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=list(range(3))
)

print(cm)


# ================================================================
# 20. SAVE CONFUSION MATRIX
# ================================================================

cm_path = os.path.join(
    EVALUATION_DIR,
    "tomato_confusion_matrix.png"
)

plt.figure(
    figsize=(8, 6)
)

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title(
    "Tomato CNN — Confusion Matrix"
)

plt.colorbar()

tick_marks = np.arange(
    len(class_names)
)

plt.xticks(
    tick_marks,
    class_names,
    rotation=45,
    ha="right"
)

plt.yticks(
    tick_marks,
    class_names
)

threshold = cm.max() / 2.0

for i in range(cm.shape[0]):

    for j in range(cm.shape[1]):

        plt.text(
            j,
            i,
            cm[i, j],
            horizontalalignment="center",
            color="white" if cm[i, j] > threshold else "black"
        )

plt.ylabel("Actual Class")

plt.xlabel("Predicted Class")

plt.tight_layout()

plt.savefig(
    cm_path,
    dpi=300,
    bbox_inches="tight"
)

plt.close()

print("✅ Confusion matrix saved:")
print(cm_path)


# ================================================================
# 21. CLASSIFICATION REPORT
# ================================================================

print("\n" + "=" * 80)
print("📋 CLASSIFICATION REPORT")
print("=" * 80)

report_text = classification_report(
    y_test,
    y_pred,
    target_names=class_names,
    digits=4,
    zero_division=0
)

print(report_text)


# Save TXT
report_txt_path = os.path.join(
    EVALUATION_DIR,
    "tomato_classification_report.txt"
)

with open(
    report_txt_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "TOMATO CNN — CLASSIFICATION REPORT\n"
    )

    f.write("=" * 70 + "\n\n")

    f.write(report_text)

print(
    "✅ Classification report saved:"
)

print(report_txt_path)


# ================================================================
# 22. SAVE CLASSIFICATION CSV
# ================================================================

report_dict = classification_report(
    y_test,
    y_pred,
    target_names=class_names,
    output_dict=True,
    zero_division=0
)

report_rows = []

for name in class_names:

    row = report_dict[name]

    report_rows.append({
        "class": name,
        "precision": row["precision"],
        "recall": row["recall"],
        "f1_score": row["f1-score"],
        "support": int(row["support"])
    })


report_csv_path = os.path.join(
    EVALUATION_DIR,
    "tomato_classification_report.csv"
)

pd.DataFrame(
    report_rows
).to_csv(
    report_csv_path,
    index=False
)

print(
    "✅ Classification CSV saved:"
)

print(report_csv_path)


# ================================================================
# 23. SAVE TEST PREDICTIONS
# ================================================================

print("\n" + "=" * 80)
print("🔮 SAVING TEST PREDICTIONS")
print("=" * 80)

prediction_rows = []

for position, original_index in enumerate(
    test_indices
):

    actual = int(
        y_test[position]
    )

    predicted = int(
        y_pred[position]
    )

    row = {
        "original_index": int(original_index),
        "actual_class": str(
            class_names[actual]
        ),
        "predicted_class": str(
            class_names[predicted]
        ),
        "confidence": float(
            confidence[position]
        ),
        "correct": bool(
            actual == predicted
        ),
        "source": str(
            source[original_index]
        )
    }

    # Add probability for each class
    for class_id, name in enumerate(
        class_names
    ):

        safe_name = str(name).lower()

        row[
            f"probability_{safe_name}"
        ] = float(
            predictions[position, class_id]
        )

    prediction_rows.append(row)


predictions_df = pd.DataFrame(
    prediction_rows
)

predictions_path = os.path.join(
    EVALUATION_DIR,
    "tomato_test_predictions.csv"
)

predictions_df.to_csv(
    predictions_path,
    index=False
)

print(
    "✅ Test predictions saved:"
)

print(predictions_path)


# ================================================================
# 24. MISCLASSIFICATION ANALYSIS
# ================================================================

print("\n" + "=" * 80)
print("🔍 MISCLASSIFICATION CHECK")
print("=" * 80)

incorrect_mask = (
    y_test != y_pred
)

incorrect_count = int(
    np.sum(incorrect_mask)
)

correct_count = int(
    np.sum(~incorrect_mask)
)

print("Correct predictions      :", correct_count)
print("Incorrect predictions    :", incorrect_count)

if incorrect_count > 0:

    print("\nMisclassification pairs:")

    mis_df = predictions_df[
        predictions_df["correct"] == False
    ]

    pair_counts = (
        mis_df.groupby(
            [
                "actual_class",
                "predicted_class"
            ]
        )
        .size()
        .reset_index(
            name="count"
        )
        .sort_values(
            "count",
            ascending=False
        )
    )

    print(
        pair_counts.to_string(
            index=False
        )
    )

else:

    print(
        "🎉 No test-set misclassifications."
    )


# ================================================================
# 25. SAVE EVALUATION SUMMARY
# ================================================================

summary = {

    "project": "Plant Disease Detection",

    "model": "Tomato CNN — MobileNetV2",

    "model_path": MODEL_PATH,

    "dataset_path": DATASET_PATH,

    "total_images": int(len(X)),

    "image_shape": [
        224,
        224,
        3
    ],

    "classes": [
        str(x)
        for x in class_names
    ],

    "class_distribution": {
        str(class_names[i]): int(
            np.sum(y == i)
        )
        for i in range(3)
    },

    "duplicate_groups": int(
        len(duplicate_groups)
    ),

    "train_images": int(
        len(train_indices)
    ),

    "validation_images": int(
        len(val_indices)
    ),

    "test_images": int(
        len(test_indices)
    ),

    "test_class_distribution": {
        str(class_names[i]): int(
            np.sum(y_test == i)
        )
        for i in range(3)
    },

    "test_accuracy": float(
        test_accuracy
    ),

    "test_accuracy_percent": float(
        test_accuracy * 100
    ),

    "per_class": {
        str(class_names[i]): {
            "precision": float(
                precision[i]
            ),
            "recall": float(
                recall[i]
            ),
            "f1_score": float(
                f1[i]
            ),
            "support": int(
                support[i]
            )
        }
        for i in range(3)
    },

    "correct_predictions": correct_count,

    "incorrect_predictions": incorrect_count,

    "seed": SEED

}


summary_path = os.path.join(
    EVALUATION_DIR,
    "tomato_evaluation_summary.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=4
    )

print(
    "\n✅ Evaluation summary saved:"
)

print(summary_path)


# ================================================================
# 26. FINAL SANITY CHECK
# ================================================================

print("\n" + "=" * 80)
print("🔐 FINAL EVALUATION SANITY CHECK")
print("=" * 80)

assert len(test_indices) == 270

assert len(y_test) == 270

assert len(y_pred) == 270

assert cm.sum() == 270

assert len(predictions_df) == 270

assert (
    correct_count +
    incorrect_count
) == 270

assert os.path.isfile(
    MODEL_PATH
)

assert os.path.isfile(
    DATASET_PATH
)

assert os.path.isfile(
    cm_path
)

assert os.path.isfile(
    report_txt_path
)

assert os.path.isfile(
    report_csv_path
)

assert os.path.isfile(
    predictions_path
)

assert os.path.isfile(
    summary_path
)

print("✅ 270 test predictions verified")
print("✅ Confusion matrix total = 270")
print("✅ Prediction CSV contains 270 rows")
print("✅ All evaluation artifacts exist")
print("✅ Final model remains unchanged")
print("✅ Dataset remains unchanged")


# ================================================================
# 27. FINAL RESULT
# ================================================================

print("\n" + "=" * 80)
print("🏁 TOMATO CNN — FINAL MODEL EVALUATION COMPLETE")
print("=" * 80)

print("\nMODEL:")
print(MODEL_PATH)

print("\nDATASET:")
print(DATASET_PATH)

print("\nTEST SET:")
print("270 images")

print("\nTEST ACCURACY:")
print(
    f"{test_accuracy * 100:.2f}%"
)

print("\nCORRECT:")
print(correct_count)

print("INCORRECT:")
print(incorrect_count)

print("\nCLASSIFICATION REPORT:")
print(report_txt_path)

print("\nCONFUSION MATRIX:")
print(cm_path)

print("\nPREDICTIONS:")
print(predictions_path)

print("\nSUMMARY:")
print(summary_path)

print("\n" + "=" * 80)
print("🎉 TOMATO CNN — EVALUATION PASS")
print("=" * 80)
```
