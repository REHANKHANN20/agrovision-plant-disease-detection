```python
# Step 3: Run Streamlit Web Application in Background
import subprocess, time

APP_PATH = '/content/drive/MyDrive/Plant Disease Detection (Computer Vision)/8th Web_Application/app.py'

# Kill any existing background instances
!pkill -f streamlit
!pkill -f cloudflared

print('🚀 Launching Streamlit App...')
subprocess.Popen([
    'streamlit', 'run', APP_PATH,
    '--server.port', '8501',
    '--server.headless', 'true',
    '--server.enableCORS', 'false',
    '--server.enableXsrfProtection', 'false'
])
time.sleep(5)
print('✅ Streamlit App is active!')
```

```python
# Step 4: Expose Live Public URL via Cloudflare Tunnel
print('🌐 Connecting Cloudflare Public Tunnel...')
print('👇 Look below for the URL ending with .trycloudflare.com (Click to open immediately! No password required!)\n')
!cloudflared tunnel --url http://127.0.0.1:8501
```
